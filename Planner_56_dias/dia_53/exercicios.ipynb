## Dia 53 — Ferramentas limitadas antes de agentes

### Situação

Uma equipe quer preparar as ferramentas de um futuro copiloto antes de conectá-lo a um LLM. Você deverá implementar um roteador pequeno que execute apenas consultas permitidas e deixe claro quando uma resposta é parcial ou uma operação falhou. O roteador será determinístico: suas decisões serão definidas por regras explícitas.

### Dados disponíveis

Use `pedidos.csv`, com oito pedidos nas colunas `id`, `tipo`, `argumento` e `falha`; `ferramenta_dados.csv`, com totais por `grupo`; e `ferramenta_documento.csv`, com `chave`, `texto` e `fonte`.

Para o tipo `dados`, consulte o grupo indicado em `argumento`. Para `documento`, consulte a chave indicada. Para `hibrida`, use o argumento como grupo e consulte também o documento de chave `prazo`. Os tipos `escrita` e `outro` são proibidos. A coluna `falha` determina a simulação: `nenhuma` executa normalmente; `timeout` faz a ferramenta documental lançar timeout; `dados_timeout` faz a ferramenta numérica lançar timeout.

### Sua tarefa

1. Implemente duas funções de leitura, uma por ferramenta, com entradas e saídas definidas e anotadas por tipo. Preserve o valor numérico e a fonte do documento retornado.
2. Implemente o roteador conforme as regras, com registro das chamadas e limite máximo de duas ferramentas por pedido, contando também tentativas que falharem.
3. Trate grupo inexistente, timeout e operação proibida com resultados identificáveis. Pedidos proibidos não podem acionar ferramentas.
4. Em um pedido híbrido cuja consulta numérica falhe, mantenha a consulta documental possível e devolva resultado parcial claramente identificado, sem inventar um total.
5. Execute os oito pedidos e escreva testes dos limites e comportamentos. Não execute comandos livres, escrita de dados ou chamadas de LLM.

### Entrega

Apresente funções, roteador, registro dos oito resultados e testes. Não é necessário construir um sistema multiagente.

### Confira seu avanço

- [ ] Todos os pedidos terminam, e nenhum executa mais de duas chamadas.
- [ ] Tipos proibidos não acionam ferramentas; valores e fontes das leituras válidas são preservados.
- [ ] Sucesso completo, resultado parcial e falha estão diferenciados.

2. Como permitir SQL arbitrário ou comandos livres mudaria o escopo e o risco do roteador?
