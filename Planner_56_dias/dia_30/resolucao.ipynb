## Dia 30 — Resíduo não é só número que sobra

### Situação

Um serviço mediu como o tempo de resposta varia com a quantidade de usuários. Uma reta pode resumir parte da relação e, ainda assim, deixar um padrão nos erros. Você deverá comparar alternativas simples e investigar se os resíduos revelam uma estrutura que o modelo não captou.

### Dados disponíveis

Use `carga.csv`, com `usuarios` e `tempo_ms`, para 12 níveis de usuários. Treine apenas com os níveis de 1 a 8 e avalie nos níveis de 9 a 12. A avaliação está fora da faixa de treino: é uma experiência didática de extrapolação, não uma amostra aleatória da mesma faixa.

### Sua tarefa

1. Separe os dados conforme os níveis de usuários antes de ajustar qualquer método.
2. Ajuste uma regressão linear e uma regressão polinomial de grau 2 usando apenas o treino. Acrescente uma referência simples que sempre preveja a média dos tempos de treino.
3. Calcule o erro absoluto médio, ou MAE, dos três métodos no treino e na avaliação, usando os mesmos registros para todos.
4. Calcule resíduos como observado menos previsto e faça gráficos contra o número de usuários, distinguindo treino e avaliação.
5. Recomende um método com base nos erros e nos padrões residuais. Explique os riscos de extrapolar a relação a partir de poucos pontos.

### Entrega

Apresente o código, a tabela de MAE, os gráficos de resíduos e uma recomendação limitada a este experimento.

### Confira seu avanço

- [ ] Os três métodos usam os mesmos recortes de treino e avaliação.
- [ ] A média de referência e os ajustes foram obtidos sem acesso aos quatro alvos de avaliação.
- [ ] Minha recomendação considera padrões residuais e extrapolação, em vez de escolher apenas pelo R² de treino.

### Interpretação

1. Um resíduo médio próximo de zero garante que um modelo descreva bem a relação?
2. Por que quatro pontos de avaliação não sustentam uma conclusão definitiva sobre o desempenho futuro?
