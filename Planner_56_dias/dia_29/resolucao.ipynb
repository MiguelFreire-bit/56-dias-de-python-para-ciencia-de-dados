## Dia 29 — Uma regressão que você consegue explicar

### Situação

Uma imobiliária quer descrever a relação entre área e aluguel em um pequeno conjunto de imóveis. Você deverá ajustar uma reta, explicar seus componentes e avaliar os limites de usá-la para estimar o aluguel de imóveis com outras áreas.

### Dados disponíveis

Use `imoveis.csv`, com oito imóveis fictícios nas colunas `area_m2` e `aluguel_reais`. Trata-se de um recorte observacional. Nesta atividade, todos os registros serão usados no ajuste didático, sem conjunto de teste. Portanto, o erro de ajuste não deve ser apresentado como evidência de desempenho em novos imóveis.

### Sua tarefa

1. Defina área como entrada `x` e aluguel como resposta `y`. Calcule a inclinação da reta por `sum((x - média_x) × (y - média_y)) / sum((x - média_x)²)` e o intercepto por `média_y - inclinação × média_x`.
2. Ajuste uma `LinearRegression` com intercepto e confira os dois coeficientes obtidos.
3. Calcule os valores ajustados e os resíduos, definidos como aluguel observado menos aluguel ajustado. Faça um gráfico dos dados com a reta.
4. Estime o aluguel para 65 m² e para 250 m². Compare essas áreas com a faixa observada e explique a diferença entre interpolar e extrapolar.

### Entrega

Apresente a equação com suas unidades, os cálculos dos coeficientes, a tabela de resíduos, o gráfico e a avaliação das duas estimativas.

### Confira seu avanço

- [ ] Os coeficientes calculados por somas e pela biblioteca diferem em menos de `1e-6`.
- [ ] A soma dos resíduos do ajuste com intercepto fica aproximadamente igual a zero.
- [ ] Diferenciei interpolação e extrapolação e não tratei o erro de treino como validação ou o coeficiente como efeito causal.

### Interpretação

1. Qual é a unidade da inclinação e o que ela resume sobre esses imóveis?
2. O intercepto permite afirmar o aluguel de um imóvel de área zero? Por quê?
