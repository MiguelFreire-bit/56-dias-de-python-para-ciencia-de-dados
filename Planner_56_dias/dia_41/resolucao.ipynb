## Dia 41 — Uma API pode falhar sem estragar os dados

### Situação

Uma rotina importa taxas de inflação de um serviço HTTP. Quando o acesso falha, ela não pode devolver uma tabela vazia como se a consulta tivesse funcionado e não houvesse dados. Você deverá implementar a leitura e testar sucessos e falhas sem depender da disponibilidade da internet.

### Dados disponíveis

Na pasta estão `ipca_original.json`, com os campos textuais `data` e `valor` da série SGS 433 do Banco Central, e `ipca_2023_2024.csv`, com o resultado de referência nas colunas `data` e `ipca_pct`. A série é o IPCA/IBGE de janeiro de 2023 a dezembro de 2024. O arquivo `respostas_http.csv` descreve quatro cenários: sucesso 200, erro 404, corpo inválido com status 200 e timeout. No sucesso, o corpo deve ser lido do JSON local; no timeout, o cliente simulado deve lançar uma exceção.

### Sua tarefa

1. Defina um cliente substituível, isto é, um objeto ou função que seu importador possa receber para obter status e corpo da resposta. Nos testes, esse cliente apenas reproduzirá os cenários locais.
2. Implemente a validação do status e do JSON. No sucesso, converta datas de dia/mês/ano e valores numéricos para o esquema do CSV de referência.
3. Escreva quatro testes de comportamento, um para cada cenário, verificando o resultado válido ou uma falha identificável. Não substitua falhas por uma saída vazia indistinguível de sucesso.
4. Confira que o sucesso produz os 24 meses únicos e os valores do CSV. Preserve a origem e o conteúdo bruto para auditoria. Se implementar acesso real opcional, defina timeout e não crie repetição ilimitada de tentativas.

### Entrega

Apresente a função de importação, o cliente simulado, os quatro testes e a conferência com o CSV. Uma chamada real não é necessária para concluir.

### Confira seu avanço

- [ ] O cenário de sucesso produz 24 meses únicos e valores numéricos, e falhas não se passam por ausência de dados.
- [ ] Os testes distinguem 404, timeout e JSON inválido; qualquer cliente real opcional tem timeout explícito.
- [ ] Mantive os dados brutos e a origem e não implementei tentativas sem limite.

2. Como o consumidor da função distinguirá falta de observações de falha de acesso?
