## Dia 55 — Receber um problema e escolher um caminho

### Situação

Um laboratório recebeu medidas químicas de vinhos e quer uma prova de conceito reproduzível para distinguir cultivares. Desta vez, cabe a você escolher a abordagem e justificar as decisões, respeitando um orçamento pequeno de comparação. A tarefa não exige atingir uma pontuação artificialmente estabelecida.

### Dados disponíveis

Use `wine.csv`, cópia local da base Wine/UCI, com 178 amostras, 13 atributos químicos, `row_id` e `target` em 0, 1 e 2. Exclua o identificador e o alvo das entradas. Não use dados externos. Reserve 25% como teste estratificado com semente 55. No desenvolvimento, use validação cruzada estratificada de quatro divisões, com embaralhamento e semente 55.

### Sua tarefa

1. Antes de comparar resultados, defina a métrica principal, o procedimento de validação e as transformações que pretende usar. Faça um diagnóstico breve apenas do conjunto de desenvolvimento.
2. Compare uma referência simples e no máximo dois tipos de modelo, com até três configurações por tipo. Coloque transformações aprendidas dentro de `Pipeline`, para que sejam ajustadas em cada treino da validação cruzada.
3. Escolha a abordagem pelo desenvolvimento, registre e congele a decisão. Ajuste-a no desenvolvimento completo e avalie uma única vez no teste.
4. Apresente a métrica principal, resultados por classe e comparação com a referência. Se a referência não for superada, analise o resultado sem expandir a busca com base no teste.
5. Indique limitações e uma melhoria ligada a uma falha observada. Organize a execução para que outra pessoa consiga reproduzi-la.

### Entrega

Apresente notebook ou script reexecutável e relatório de até 400 palavras com problema, protocolo, decisão, resultados e limites. Registre sementes e versões das bibliotecas usadas.
