## Dia 33 — Um pipeline mínimo de classificação

### Situação

Um laboratório quer distinguir três cultivares de vinho a partir de medidas químicas. Você deverá montar um fluxo simples de classificação e verificar se ele oferece informação além de uma previsão constante. O objetivo é identificar cultivar, e não qualidade ou preferência de sabor.

### Dados disponíveis

Use `wine.csv`, cópia local da base Wine/UCI distribuída pelo scikit-learn. Há 178 amostras, 13 medidas químicas, o identificador `row_id` e o alvo `target`, com classes 0, 1 e 2. As medidas têm escalas diferentes. Não atribua unidades que não estejam documentadas.

### Sua tarefa

1. Separe as 13 medidas como entradas, excluindo identificador e alvo. Reserve 25% para teste com estratificação pelo alvo e `random_state=33`.
2. Ajuste no treino um `Pipeline` com `StandardScaler` e `LogisticRegression(max_iter=2000)`. Ajuste também um `DummyClassifier(strategy="most_frequent")` como referência.
3. Avalie os dois métodos no mesmo teste usando acurácia, macro-F1 e matriz de confusão. Informe a ordem das classes na matriz.
4. Obtenha as probabilidades por classe do modelo e confira suas somas por linha. Examine até dois casos classificados incorretamente. Se não houver erros, discuta o que o teste pequeno ainda não permite assegurar.
