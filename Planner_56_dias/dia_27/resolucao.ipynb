## Dia 27 — Dez chances de anunciar uma descoberta

### Situação

Um experimento avaliou dez métricas primárias. O relatório pretende destacar qualquer resultado com p-valor abaixo de 0,05, sem considerar que foram feitas dez comparações. Você deverá revisar as decisões e explicar o risco de analisar cada resultado isoladamente.

### Dados disponíveis

O arquivo `testes.csv` contém `metrica` e `p_valor`. Considere válidos os dez testes individuais e defina a família como o conjunto completo desses testes. O nível de erro familiar desejado é 0,05. Apenas no cálculo teórico adicional, suponha que os dez testes sejam independentes e que todas as hipóteses nulas sejam verdadeiras.

### Sua tarefa

1. Identifique quais testes seriam rejeitados usando 0,05 separadamente em cada um.
2. Aplique Bonferroni com limiar `0,05/10`. Calcule também os p-valores ajustados, multiplicando cada p-valor por dez e limitando o resultado a 1.
3. No cenário teórico de dez testes independentes sob hipóteses nulas verdadeiras, calcule `1 - (1 - 0,05)^10`, a chance de pelo menos uma rejeição indevida sem a correção.
4. Compare as decisões e escreva uma crítica ao relatório, distinguindo erro familiar de nível individual.
