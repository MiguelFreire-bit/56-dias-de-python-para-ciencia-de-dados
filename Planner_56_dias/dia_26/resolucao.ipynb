## Dia 26 — Poucas contagens exigem cuidado

### Situação

Uma equipe sorteou 20 visitantes entre duas páginas para realizar um pequeno teste de conversão. Algumas combinações de página e resultado têm poucas pessoas. Você deverá comparar as proporções e justificar um teste adequado ao tamanho da tabela.

### Dados disponíveis

O arquivo `experimento.csv` contém `grupo`, `converteu` e `nao_converteu`. A teve uma conversão em dez visitantes; B teve cinco em dez. Cada visitante pertence a apenas um grupo e aparece uma única vez. Use teste bilateral de associação com nível de significância de 0,05.

### Sua tarefa

1. Monte uma tabela 2 × 2 de contagens, identificando a ordem dos grupos e dos resultados.
2. Calcule as proporções de conversão e a diferença B menos A, expressa também em pontos percentuais.
3. Calcule as frequências esperadas sob independência a partir dos totais de linhas e colunas.
4. Execute o teste exato de Fisher bilateral. Use o tamanho das células para explicar a escolha e escreva uma conclusão que diferencie a diferença observada da força da evidência.

### Entrega

Apresente tabelas observada e esperada, proporções, diferença, teste e um parágrafo de interpretação.

### Confira seu avanço

- [ ] Declarei a orientação da tabela e mantive a diferença como B menos A.
- [ ] Executei o teste com contagens, sem fornecer percentuais como se fossem números de pessoas.
- [ ] Não tratei ausência de evidência como equivalência nem prometi generalização ampla a partir de 20 visitantes.

### Interpretação

1. Por que a aproximação qui-quadrado pode ser frágil quando as frequências esperadas são pequenas?
2. A diferença pontual entre as proporções basta para declarar uma página superior?
