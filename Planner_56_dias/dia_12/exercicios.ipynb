## Dia 12 — Um JOIN que não inventa clientes

### Situação

Um gestor quer saber quantos clientes estão cadastrados em cada cidade e qual é o valor dos pedidos desses clientes. Um mesmo cliente pode ter vários pedidos, e há um pedido cujo cliente não existe no cadastro. Você deverá produzir os resumos sem contar um cliente várias vezes nem esconder esse problema cadastral.

### Dados disponíveis

Use `clientes.csv`, com `cliente_id` e `cidade`, e `pedidos.csv`, com `pedido_id`, `cliente_id` e `valor`. Há quatro clientes e quatro pedidos. O código C9 aparece nos pedidos, mas não no cadastro. Os valores dos pedidos estão em reais. Use SQLite, disponível na biblioteca padrão do Python pelo módulo `sqlite3`.

### Sua tarefa

1. Carregue os dois arquivos em tabelas SQLite e confira a quantidade de registros de cada uma.
2. Escreva uma consulta que parta do cadastro e use `LEFT JOIN` para manter também os clientes sem pedidos. Apresente, por cidade, o número de clientes distintos e o valor total dos pedidos associados. Para o resumo de pedidos, um cliente sem compra contribui com zero reais, sem que seja criado um pedido fictício.
3. Faça outra consulta para listar os pedidos sem cliente correspondente no cadastro.
4. Confira o valor total de todos os pedidos. Demonstre que ele é igual ao valor associado a clientes cadastrados somado ao valor dos pedidos sem cadastro.

### Entrega

Apresente o código de carga, as consultas, suas tabelas de saída, a conferência dos totais e uma explicação do que cada linha representa antes e depois da junção.

### Confira seu avanço

- [ ] Clientes sem pedidos continuam representados, sem ganhar pedidos artificiais.
- [ ] Contei clientes pela chave distinta e identifiquei explicitamente os pedidos sem cadastro.
- [ ] O valor associado aos clientes mais o valor dos pedidos sem cadastro confere com o total original.

### Interpretação

1. Por que `COUNT(*)` após uma junção pode contar pedidos ou linhas resultantes, em vez de clientes?
2. Que informações um `INNER JOIN` esconderia neste problema?
