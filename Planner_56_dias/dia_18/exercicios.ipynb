## Dia 18 — Amostra grande não cura seleção ruim

### Situação

Um serviço atende de manhã e à noite. Para estimar o tempo médio de atendimento do cadastro inteiro, um gestor escolheu os seis primeiros registros da manhã. Você deverá comparar essa seleção com duas alternativas aleatórias e avaliar como a escolha dos participantes influencia a estimativa.

### Dados disponíveis

O arquivo `cadastro.csv` contém 16 atendimentos nas colunas `id`, `turno` e `tempo_min`. Os IDs 1 a 8 são da manhã, e os IDs 9 a 16 são da noite. Neste exercício, os 16 registros formam a população finita de interesse, conhecida integralmente apenas para permitir a comparação didática.

### Sua tarefa

1. Calcule o tempo médio dos 16 atendimentos, que será a referência conhecida do exercício.
2. Forme a amostra por conveniência escolhendo os IDs 1 a 6.
3. Forme uma amostra aleatória simples de seis IDs, sem reposição, com semente 180. Em seguida, forme uma amostra estratificada: selecione três atendimentos da manhã e três da noite, sem reposição, usando um gerador com semente 181.
4. Para cada amostra, informe os IDs escolhidos, a quantidade por turno, a média e o erro em relação à média da população. Defina erro como média amostral menos média da população.
5. Recomende uma estratégia de seleção e explique por que a média simples da amostra estratificada é adequada neste cadastro, em que os dois turnos têm o mesmo tamanho.

### Entrega

Inclua código, tabela das seleções e dos erros e um parágrafo de recomendação. Use `numpy.random.default_rng` para tornar os sorteios reproduzíveis.

### Confira seu avanço

- [ ] Cada amostra tem seis IDs distintos, e a estratificada tem três de cada turno.
- [ ] Usei a média da população como referência didática, reconhecendo que ela geralmente não seria conhecida em uma pesquisa real.
- [ ] Não concluí que um método é sempre superior apenas porque um sorteio específico produziu erro menor.

### Interpretação

1. Qual é a diferença entre erro amostral e viés de seleção neste problema?
2. Se os turnos tivessem tamanhos diferentes, que cuidado seria necessário ao selecionar três atendimentos de cada um?
