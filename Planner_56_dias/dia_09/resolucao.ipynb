## Dia 9 — Associação não é explicação causal

### Situação

Uma escola registrou as horas de estudo declaradas e as notas de oito estudantes. A direção quer saber se os dois valores apresentam associação. Você deverá analisar essa relação e mostrar por que uma associação observada não demonstra que aumentar as horas de estudo, por si só, produzirá uma mudança específica na nota.

### Dados disponíveis

O arquivo `estudo.csv` contém `horas` e `nota`. As horas são 1, 2, 3, 4, 5, 6, 7 e 8; as notas correspondentes são 4, 5, 5, 6, 7, 7, 8 e 6. Cada linha representa um estudante. Ninguém foi sorteado para estudar por determinado número de horas: trata-se de um levantamento observacional.

### Sua tarefa

1. Faça um gráfico de dispersão das horas e notas, identificando os eixos.
2. Calcule a covariância amostral, usando divisor `n-1`, e a correlação de Pearson.
3. Crie uma coluna com o tempo de estudo em minutos e recalcule as duas medidas usando essa coluna no lugar de horas. Compare os resultados e suas unidades.
4. Explique a associação observada e indique pelo menos um fator não medido que poderia influenciar tanto o tempo de estudo quanto a nota.

### Entrega

Apresente o gráfico, a tabela das medidas nas duas escalas e um parágrafo de interpretação.

### Confira seu avanço

- [ ] A correlação está entre -1 e 1, e a conversão de horas para minutos usa fator 60.
- [ ] Demonstrei numericamente como a mudança de escala afeta a covariância e a correlação.
- [ ] A conclusão menciona um possível fator de confundimento e não afirma uma relação causal comprovada.

### Interpretação

1. Como o último estudante se diferencia da tendência sugerida pelos demais pontos?
2. Mesmo uma correlação alta garantiria uma previsão precisa para um novo estudante? Por quê?
