## Dia 38 — Escolher o limiar pelo custo

### Situação

Uma equipe precisa definir a partir de qual pontuação um caso deve gerar alerta. Para a operação, deixar de sinalizar um positivo custa cinco vezes mais do que emitir um alerta falso. Você recebeu resultados de validação e deverá escolher um limiar considerando esse custo.

### Dados disponíveis

Use `validacao.csv`, com 12 casos fictícios nas colunas `id`, `real` e `score`. `real` vale 0 ou 1, e `score` está entre 0 e 1. Avalie os limiares 0,2, 0,4, 0,6 e 0,8. Classifique como positivo quando `score >= limiar`. O custo total é `FP + 5 × FN`. Não há conjunto de teste nesta atividade.

### Sua tarefa

1. Gere as classificações para cada um dos quatro limiares, incluindo corretamente os casos exatamente no limite.
2. Calcule matriz de confusão, custo total, precisão e recall de cada limiar. Declare uma convenção para métricas com denominador zero.
3. Escolha o menor custo; em caso de empate, prefira o limiar maior.
4. Explique os ganhos e perdas da escolha e indique que ainda seria necessária uma avaliação independente para estimar seu desempenho final.

### Entrega

Apresente a tabela dos quatro limiares e uma justificativa da decisão e de seus limites.

### Confira seu avanço

- [ ] Usei a regra `>=`, inclusive nas pontuações iguais ao limiar.
- [ ] Os custos foram aplicados às contagens corretas, e a escolha usou somente a validação.
- [ ] Não tratei o score como probabilidade calibrada sem evidência nem o custo escolhido como resultado de teste independente.

### Interpretação

1. Como aumentar o custo dos falsos negativos poderia alterar a escolha?
2. Por que uma prevalência diferente em produção poderia mudar a utilidade desse limiar?
