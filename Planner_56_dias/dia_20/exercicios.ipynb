## Dia 20 — Dispersão de taxas perto de zero

### Situação

Um relatório usa o coeficiente de variação para comparar qualquer coluna numérica. Você recebeu a tarefa de avaliar se essa escolha é adequada para taxas mensais de inflação, que podem ser negativas ou próximas de zero. A comparação deverá considerar também medidas de dispersão absoluta.

### Dados disponíveis

Use `ipca_2023_2024.csv`, com as colunas `data` e `ipca_pct` e 24 meses de 2023 e 2024. Os dados são do IPCA do IBGE, distribuídos pelo Banco Central na série SGS 433. Para este exercício, defina o coeficiente de variação como desvio padrão dividido pela média, sem tomar o valor absoluto da média. Use `ddof=0` nas descrições anuais.

### Sua tarefa

1. Separe os anos e calcule, em cada um, média, desvio padrão, intervalo interquartil e coeficiente de variação. Use interpolação linear nos quartis.
2. Crie uma cópia artificial das taxas subtraindo 0,40 ponto percentual de cada valor. Não altere a série original e não apresente a transformação como IPCA observado.
3. Repita os resumos na série artificial e compare como a mudança da origem da escala afeta cada medida. Se uma média for zero, registre o coeficiente de variação como indefinido.
4. Escolha uma forma de comunicar a dispersão dessas taxas e justifique por que ela é adequada ao contexto.

### Entrega

Apresente a tabela dos dois anos nas séries original e transformada e sua recomendação de medida.

### Confira seu avanço

- [ ] Diferenciei taxa percentual de mudança em ponto percentual.
- [ ] Identifiquei a série artificial e examinei a sensibilidade do coeficiente de variação à média e à origem da escala.
- [ ] Não tratei o coeficiente de variação como regra universal nem os meses como amostra independente para inferência.

### Interpretação

1. Qual medida mantém uma interpretação útil quando a média se aproxima de zero?
2. Por que subtrair uma constante não deveria mudar uma medida de dispersão absoluta baseada nas diferenças entre valores?
