## Dia 32 — Aprender a separar antes de transformar

### Situação

Uma equipe padronizou todas as medidas de flores antes de separar treino e teste. Você deverá comparar esse procedimento com um fluxo correto e identificar que informação do teste foi usada indevidamente, mesmo que as acurácias finais sejam parecidas.

### Dados disponíveis

Use `iris.csv`, cópia local da base Iris/UCI distribuída pelo scikit-learn. São 150 plantas, quatro medidas em centímetros, `row_id` e `target`. O alvo identifica a espécie: 0 para setosa, 1 para versicolor e 2 para virginica. `row_id` identifica a linha e não deve ser usado para prever a espécie.

### Sua tarefa

1. Separe as quatro medidas como entradas e `target` como alvo. Faça um corte de 70% para treino e 30% para teste, estratificado pelo alvo, com `random_state=32`. Guarde os índices para reutilizar o mesmo corte.
2. No fluxo demonstrativo incorreto, ajuste `StandardScaler` com todas as entradas e depois use os índices guardados para separar os dados transformados. Treine KNN com cinco vizinhos apenas nas linhas de treino.
3. No fluxo correto, ajuste outro `StandardScaler` somente no treino. Transforme o teste com esse ajuste e treine outro KNN de cinco vizinhos no treino transformado.
4. Compare as médias aprendidas pelos padronizadores e as acurácias no mesmo teste. Identifique explicitamente o fluxo global como demonstração de vazamento, e não como alternativa válida.

### Entrega

Apresente os dois fluxos identificados, as médias dos padronizadores, as acurácias e o diagnóstico do procedimento.

### Confira seu avanço

- [ ] Nem `row_id` nem `target` aparecem entre as entradas do modelo.
- [ ] No fluxo correto, os ajustes acessam somente treino; no teste, usei apenas transformação e previsão.
- [ ] Expliquei o vazamento independentemente de qual fluxo obteve maior acurácia.

### Interpretação

1. Por que resultados parecidos não tornam válido o procedimento que usou informações do teste?
2. Como um `Pipeline` ajudaria a manter essa separação dentro da validação cruzada?
