## Dia 36 — O futuro não pode treinar o passado

### Situação

Uma equipe embaralhou meses para avaliar uma previsão de inflação e obteve um resultado aparentemente bom. Você deverá construir uma avaliação que respeite o tempo e a informação disponível no momento de cada previsão.

### Dados disponíveis

Use `ipca_2023_2024.csv`, com `data` e `ipca_pct`, do IPCA/IBGE distribuído pelo Banco Central na série SGS 433. São 24 meses de 2023 e 2024. Para prever a taxa de um mês, a única entrada será a taxa publicada do mês anterior. Considere que cada previsão é feita depois da publicação dessa taxa anterior.

### Sua tarefa

1. Ordene as datas e crie a entrada defasada de um mês. Janeiro de 2023 não terá entrada disponível e não poderá ser usado como alvo de treino.
2. Treine uma regressão linear com os alvos de fevereiro a dezembro de 2023. Congele o ajuste antes de avaliar 2024.
3. Para cada alvo de janeiro a dezembro de 2024, preveja com a taxa anterior realmente disponível naquela origem. Compare com a persistência, que simplesmente repete essa taxa anterior.
4. Calcule MAE para os dois métodos nos mesmos 12 meses e produza um gráfico de observado e previsões. Explique por que isso é uma sequência de previsões de um passo, e não uma previsão simultânea de todo o ano.

### Entrega

Apresente os cortes temporais, a tabela das 12 previsões, métricas, gráfico e uma avaliação das limitações.

### Confira seu avanço

- [ ] Nenhuma entrada utiliza a taxa do próprio mês previsto ou de um mês posterior.
- [ ] Os dois métodos são comparados nos mesmos 12 meses, sem ajuste dos coeficientes em 2024.
- [ ] Identifiquei a avaliação como previsão de um passo com origens sucessivas.

### Interpretação

1. Como embaralhar os meses mudaria a informação disponível e o significado dessa avaliação?
2. Que limites precisam acompanhar uma regressão treinada com apenas 11 alvos?
