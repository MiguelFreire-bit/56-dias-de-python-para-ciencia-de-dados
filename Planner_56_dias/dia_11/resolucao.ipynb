## Dia 11 — Ausência também conta uma história

### Situação

Uma pesquisa convidou oito pessoas a informar a renda, mas apenas cinco responderam. A chefia pediu um único valor para representar a renda média dos oito convidados. Você deverá mostrar o que pode ser calculado diretamente e como a falta de respostas limita essa solicitação.

### Dados disponíveis

O arquivo `rendas.csv` contém `id`, `respondeu` e `renda_reais`. Uma renda vazia representa uma não resposta. Renda zero seria válida, mas não foi observada. Para uma comparação hipotética, considere que cada renda ausente poderia estar entre R$ 1.000 e R$ 6.000. Esses limites são suposições do exercício, e não informações recuperadas sobre as pessoas.

### Sua tarefa

1. Confira quantas pessoas foram convidadas e quantas responderam. Calcule a taxa de não resposta e a média apenas das rendas informadas.
2. Em uma cópia dos dados, atribua R$ 1.000 a cada renda ausente e calcule a média dos oito convidados. Em outra cópia, repita com R$ 6.000. Identifique esses resultados como extremos do cenário hipotético.
3. Calcule também a média obtida ao preencher as ausências pela média dos respondentes, sem alterar o arquivo original.
4. Compare os resultados e explique por que o preenchimento, sozinho, não demonstra que as cinco pessoas representam as oito.

### Entrega

Apresente o código, uma tabela que separe resultado observado e cenários hipotéticos e um parecer sobre o pedido da chefia.

### Confira seu avanço

- [ ] A média observada usa cinco rendas, e a taxa de não resposta considera oito convidados.
- [ ] Os valores atribuídos nos cenários não são apresentados como rendas conhecidas.
- [ ] Minha conclusão discute a possibilidade de viés de não resposta, sem presumir que as ausências sejam aleatórias.

### Interpretação

1. Por que preencher as ausências pela média observada não elimina necessariamente o viés?
2. Que hipótese sobre respondentes e não respondentes seria necessária para generalizar a média observada?
