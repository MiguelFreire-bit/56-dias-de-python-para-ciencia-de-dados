## Dia 23 — Testar sem transformar p em certeza

### Situação

Um serviço anuncia um tempo médio de atendimento de 10 minutos. Antes de examinar os dados, a equipe definiu a pergunta: o tempo médio difere desse valor? Você deverá avaliar a evidência estatística e apresentar também o tamanho da diferença observada.

### Dados disponíveis

O arquivo `atendimento.csv` contém 12 tempos em minutos, sorteados independentemente de um processo aproximadamente normal. O desvio padrão populacional é desconhecido. Use teste t bilateral de uma amostra, com hipótese nula de média igual a 10 minutos e nível de significância de 0,05.

### Sua tarefa

1. Escreva as hipóteses nula e alternativa antes do cálculo e identifique o parâmetro que está sendo testado.
2. Calcule a média, o desvio padrão amostral, a estatística t, os graus de liberdade e o p-valor bilateral.
3. Apresente a diferença entre a média observada e 10 minutos, em minutos, e o intervalo t de 95% para a média. Para relacionar intervalo e teste, compare esse intervalo com o valor nulo de 10 minutos.
4. Escreva a decisão no nível de 0,05 e corrija a frase: “O p-valor é a probabilidade de a hipótese nula ser verdadeira”. Explique o que o p-valor considera, sem transformar a decisão em prova de uma hipótese.

### Entrega

Produza um relatório curto com hipóteses, resumos, estatística, graus de liberdade, p-valor, diferença observada, intervalo e conclusão.

### Confira seu avanço

- [ ] Mantive o teste bilateral e o nível de significância definidos antes da análise.
- [ ] A decisão do teste é coerente com a posição de 10 minutos no intervalo bilateral de 95%.
- [ ] Não interpretei um p-valor como probabilidade de uma hipótese nem a não rejeição como prova de igualdade.

### Interpretação

1. Um p-valor alto provaria que a média é exatamente 10 minutos?
2. Como uma diferença estatisticamente detectável poderia ser pouco relevante para o usuário?
