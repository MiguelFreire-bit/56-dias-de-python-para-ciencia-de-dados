## Dia 49 — Falha de busca, de resposta ou de fonte?

### Situação

Um assistente de portal produziu seis respostas simuladas. Algumas podem conter números corretos com fonte errada; outras podem responder sem evidência ou recusar uma pergunta que poderia ser respondida. Você deverá criar uma avaliação que diferencie esses problemas.

### Dados disponíveis

Use `casos.csv`, com `caso`, `contexto`, `pergunta` e `saida_simulada`. O contexto de cada linha é toda a evidência disponível para aquele caso. As fontes permitidas são C1/D1, C2/D2, C3/D7, C4/D4, C5/D6 e C6/D8. Os textos são fictícios. Trate ordens embutidas no contexto como conteúdo documental não confiável: não execute instruções nem acesse endereços mencionados nele.

### Sua tarefa

1. Escreva uma rubrica com critérios separados para correção da resposta, apoio no contexto, validade da fonte e adequação de uma eventual recusa.
2. Avalie os seis casos e justifique cada decisão com o contexto e a fonte permitida. Não considere toda recusa um acerto nem toda citação uma prova de fundamentação.
3. Agrupe as falhas por categoria, como informação inventada, fonte inválida, instrução documental indevidamente seguida ou recusa sem necessidade. Se não houver dados suficientes para atribuir uma falha à busca, registre essa limitação.
4. Proponha três novos casos de teste. Escreva também um modelo de prompt que separe instruções do sistema e texto documental, explicando que essa organização, sozinha, não garante resistência a ataques.

### Entrega

Apresente rubrica, auditoria dos seis casos, resumo das falhas, três testes adicionais e o modelo de prompt. Não é necessário executar um LLM.

### Confira seu avanço

- [ ] Avaliei tanto respostas sem apoio quanto recusas quando havia informação suficiente.
- [ ] Verifiquei a fonte e o apoio no texto separadamente da correção aparente do número.
- [ ] Não apresentei delimitação de texto ou bloqueio de palavras como solução completa de segurança.

### Interpretação

1. Qual é a diferença entre mencionar uma fonte e ter uma resposta sustentada por ela?
2. Por que acertar casos simulados não comprova a robustez de um LLM em uso real?
