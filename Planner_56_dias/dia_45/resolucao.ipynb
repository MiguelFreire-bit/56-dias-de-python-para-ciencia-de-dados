## Dia 45 — Acertar um trecho não é recuperar todos

### Situação

Uma equipe comparou duas buscas e chamou todos os indicadores de “acurácia do RAG”. Você deverá separar as métricas de recuperação e mostrar que encontrar algum documento relevante é diferente de recuperar todo o material necessário.

### Dados disponíveis

O arquivo `ranking.csv` contém `consulta`, `relevantes`, `ranking_A` e `ranking_B`. Os IDs nas células são separados por ponto e vírgula. Os rankings já estão prontos. Considere os três primeiros documentos de cada busca. Q1, Q2 e Q3 têm documentos relevantes; Q4 não tem resposta no acervo e sua lista de relevantes está vazia.

### Sua tarefa

1. Converta as listas de IDs e conte, para cada pergunta respondível, quantos documentos relevantes aparecem entre os três primeiros resultados.
2. Calcule `Hit@3`, que vale 1 quando ao menos um relevante é recuperado; `Recall@3`, que divide os relevantes recuperados pelo total de relevantes; e `Precision@3`, que divide os relevantes recuperados por 3.
3. Calcule a média simples de cada métrica nas perguntas Q1 a Q3, separadamente por busca.
4. Registre Q4 em uma análise separada: o recall é indefinido porque não há relevantes. Explique por que o retorno de candidatos nessa pergunta não demonstra que o sistema consiga reconhecer a ausência de resposta.

### Entrega
