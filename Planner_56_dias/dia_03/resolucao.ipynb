## Dia 3 — Uma entrega muito demorada

### Situação

Uma empresa quer informar a seus clientes quanto tempo uma entrega costuma levar. Foram observadas nove entregas, mas uma delas demorou muito mais do que as demais. A empresa ainda não sabe explicar essa demora e pediu que você avaliasse como esse caso influencia o resumo dos tempos.

### Dados disponíveis

O arquivo `entregas.csv` contém a coluna `minutos`, com os tempos 18, 20, 21, 22, 22, 23, 24, 25 e 90. Para esta atividade, todos os nove registros são válidos. A análise sem o valor 90 será apenas uma comparação de sensibilidade: ela não autoriza apagar o registro original.

### Sua tarefa

1. Calcule média, mediana e moda usando as nove entregas.
2. Escolha uma dessas medidas e mostre seu cálculo explicitamente, sem depender apenas do resultado de uma função pronta. Compare o cálculo com o resultado da biblioteca.
3. Crie uma cópia dos dados sem a entrega de 90 minutos e recalcule as três medidas. Mantenha a análise completa como resultado principal.
4. Apresente as duas análises lado a lado e escolha uma medida para comunicar o tempo típico de entrega. Justifique sua escolha com base no comportamento observado.

### Entrega

Inclua código, memória de cálculo de uma medida, tabela comparativa e um parágrafo com sua recomendação.

### Confira seu avanço

- [ ] A análise principal usa nove entregas, e a comparação adicional usa oito.
- [ ] O cálculo explícito e o da biblioteca diferem em menos de `1e-8`.
- [ ] Minha recomendação considera a influência da entrega demorada e não a classifica automaticamente como erro.

### Interpretação

1. Que aspecto dos tempos é resumido pela média e que aspecto é resumido pela mediana?
2. Que informação você pediria à empresa antes de recomendar a exclusão da entrega de 90 minutos?
