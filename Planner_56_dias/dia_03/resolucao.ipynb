## Resolução

Os arquivos de dados estão nesta pasta. Se necessário, confira o diretório de execução com `Path.cwd()`. Acrescente células conforme precisares.

```python
# 1. Calcule média, mediana e moda usando as nove entregas.

import pandas as pd

entregas = pd.read_csv('entregas.csv')
minutos = entregas['minutos']

print(
    f'Média:   {minutos.mean():.2f} minutos\n'
    f'Mediana: {minutos.median():.2f} minutos\n'
    f'Moda:    {minutos.mode().tolist()} minutos'
)
```

```python
# 2. Escolha uma dessas medidas e mostre seu cálculo explicitamente, 
# sem depender apenas do resultado de uma função pronta. Compare o cálculo com o resultado da biblioteca.

# vou fazer para as três

# ====== Média ======

# Média = soma dos valores / quantidade de valores.
# O erro absoluto é a diferença entre o cálculo e o resultado do pandas.

media = (18 + 20 + 21 + 22 + 22 + 23 + 24 + 25 + 90) / len([18, 20, 21, 22, 22, 23, 24, 25, 90])
media_pandas = minutos.mean()

erro_media = abs(media - media_pandas)

print(f'Média calculada: {media} minutos')
print(f'Média pelo pandas: {media_pandas} minutos')
print(f'Erro absoluto: {erro_media:.10f}')
print(f'Erro menor que 1e-8? {erro_media < 1e-8}')
print("-"*20)
# ====== Mediana ====== 

valores_ordenados = sorted(minutos)
tamanho = len(valores_ordenados)

print(f'Quantidade de valores: {tamanho}')
print(f'Valores ordenados: {valores_ordenados}')


# Ímpar: índice central = (tamanho - 1) // 2
# Par: mediana = média dos valores nos índices:
#      (tamanho // 2 - 1) e (tamanho // 2)

mediana = valores_ordenados[(tamanho - 1) // 2]

print(f'Mediana: {mediana} minutos')
print("-"*20)


# ====== Moda ======

# A moda é o valor que aparece com maior frequência.
# Quando dois ou mais valores têm a mesma frequência máxima,
# eles podem ser modas, desde que nem todos tenham a mesma frequência.
# Se todos os valores têm a mesma frequência, o conjunto é amodal.

frequencias = minutos.value_counts()

quantidade_de_repeticao_maior = int(frequencias.max())

candidatos = frequencias[
    frequencias == quantidade_de_repeticao_maior
].index.tolist()

if frequencias.nunique() == 1:
    moda = []
    print('Moda: conjunto amodal, pois todos os valores têm a mesma frequência.')
else:
    moda = candidatos
    print(f'Moda: {moda} minutos')
    print(list(minutos))
    print(f'Frequência da moda: {quantidade_de_repeticao_maior} vezes')
print("-"*20)    
```

```python
# 3. Crie uma cópia dos dados sem a entrega de 90 minutos e recalcule as três medidas. 
# Mantenha a análise completa como resultado principal.

entregas_sem_90 = entregas[entregas['minutos'] != 90].copy()

minutos_sem_90 = entregas_sem_90['minutos']

print(
    f'Média:   {minutos_sem_90.mean():.2f} minutos\n'
    f'Mediana: {minutos_sem_90.median():.2f} minutos\n'
    f'Moda:    {minutos_sem_90.mode().tolist()} minutos'
)
```

```python
# 4. Apresente as duas análises lado a lado e escolha uma medida para comunicar o tempo típico de entrega.
#  Justifique sua escolha com base no comportamento observado.

# ====== Comparação das análises ======

comparacao = pd.DataFrame({
    'Medida': ['Média', 'Mediana', 'Moda'],
    'Com as 9 entregas': [
        minutos.mean(),
        minutos.median(),
        minutos.mode().tolist(),
    ],
    'Sem a entrega de 90 min': [
        minutos_sem_90.mean(),
        minutos_sem_90.median(),
        minutos_sem_90.mode().tolist(),
    ],
})

display(comparacao)
```

Para comunicar o tempo típico das entregas observadas, eu escolheria a mediana de 22 minutos. Ao comparar os resultados com e sem a entrega de 90 minutos, a média mudou de 29,44 para 21,88 minutos, enquanto a mediana permaneceu em 22 minutos. Isso mostra que, nesse conjunto, a mediana foi menos afetada pela entrega demorada. A análise principal mantém as nove entregas, pois ainda não sabemos por que uma delas demorou mais.
