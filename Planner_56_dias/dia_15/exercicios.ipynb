## Dia 15 — A condição muda a probabilidade

### Situação

Uma loja pesquisou o canal de origem e a realização de compra de 200 pessoas. Um analista confundiu duas perguntas: a chance de comprar entre quem veio por e-mail e a chance de ter vindo por e-mail entre quem comprou. Você deverá calcular essas probabilidades e mostrar por que elas não usam necessariamente o mesmo denominador.

### Dados disponíveis

Use `pesquisa.csv`, com `canal`, `comprou` e `pessoas`. A coluna `comprou` vale 1 para quem comprou e 0 para quem não comprou. Cada linha reúne uma quantidade de pessoas, indicada em `pessoas`. Os grupos não se sobrepõem e somam 200 participantes. Considere a seleção de uma pessoa com igual chance entre esses 200 participantes.

### Sua tarefa

1. Organize uma tabela de contagens por canal e realização de compra, incluindo os totais.
2. Calcule a probabilidade de compra, a probabilidade conjunta de vir por e-mail e comprar, a probabilidade de compra dado o canal e-mail e a probabilidade de canal e-mail dado que houve compra.
3. Para cada cálculo, mostre qual grupo fornece o numerador e qual fornece o denominador.
4. Verifique se canal e compra são independentes na tabela observada, comparando uma probabilidade conjunta com o produto das marginais correspondentes. Não execute um teste de hipótese nesta atividade.

### Entrega

Apresente a tabela de contagens, os quatro cálculos e uma conclusão sobre independência no conjunto observado.

### Confira seu avanço

- [ ] Todas as probabilidades ficam entre 0 e 1, e as probabilidades das quatro combinações somam 1.
- [ ] Cada probabilidade condicional usa como denominador o grupo definido pela condição.
- [ ] A verificação de independência compara quantidades compatíveis e não generaliza automaticamente para outras pessoas.

### Interpretação

1. Por que inverter a condição pode mudar o valor da probabilidade?
2. Uma relação de independência nesta tabela provaria independência em toda a população da loja?
