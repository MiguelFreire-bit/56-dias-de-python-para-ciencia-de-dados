## Dia 46 — Oito em dez não encerra a avaliação

### Situação

Um assistente acertou oito de dez perguntas em um piloto. Em uma avaliação separada e maior, acertou 80 de 100. As proporções são iguais, mas a quantidade de evidência não é. Você deverá comunicar a incerteza de cada estimativa sem declarar equivalência entre sistemas.

### Dados disponíveis

Use `avaliacao.csv`, com `amostra`, `acertos` e `total`. Para este exercício, as perguntas de cada avaliação foram sorteadas independentemente da mesma população-alvo e julgadas com uma rubrica fixa. Calcule intervalos de Wilson de 95% com `z=1,96`. Para proporção observada `p` e tamanho `n`, use centro `(p + z²/(2n)) / (1 + z²/n)` e margem `z × sqrt(p(1-p)/n + z²/(4n²)) / (1 + z²/n)`.

### Sua tarefa

1. Calcule a proporção de acertos de cada avaliação usando `p = acertos/total`.
2. Calcule centro, margem, limites e largura do intervalo de Wilson para cada uma.
3. Faça um gráfico com as estimativas pontuais e seus intervalos e compare a precisão das estimativas.
4. Explique como a interpretação mudaria se várias perguntas fossem apenas paráfrases de uma mesma questão e, portanto, dependentes.

### Entrega

Apresente os cálculos reproduzíveis, a tabela dos intervalos, o gráfico e uma nota sobre a qualidade da avaliação.
