## Dia 37 — Acurácia alta pode esconder falhas

### Situação

Uma equipe avalia dois sistemas de detecção de fraude e anuncia apenas suas acurácias. Como fraudes são raras, um sistema pode acertar muitas transações sem detectar os casos de maior interesse. Você deverá reconstruir as matrizes de confusão e avaliar essa comunicação.

### Dados disponíveis

O arquivo `alertas.csv` contém `classe_real`, `previsao_A`, `previsao_B` e `quantidade`. A classe positiva, 1, representa fraude; 0 representa ausência de fraude. Cada linha reúne eventos com a mesma combinação de resultados. Portanto, `quantidade` deve ser usada como frequência, totalizando 100 eventos, e não apenas quatro observações. As previsões já estão prontas.

### Sua tarefa

1. Construa a matriz de confusão de cada sistema, ponderando as linhas por `quantidade`. Identifique verdadeiros e falsos positivos e negativos.
2. Calcule acurácia, precisão, recall e F1 para a classe fraude.
3. Se um sistema não prever nenhum positivo, registre que o denominador da precisão é zero e adote a convenção de reportar precisão igual a zero. Declare também a convenção usada para F1 quando seu denominador for zero.
4. Informe a prevalência de fraude e avalie os sistemas considerando o objetivo de encontrar fraudes, em vez de se apoiar apenas na acurácia.

### Entrega

Apresente matrizes, tabela de métricas e uma crítica ao anúncio da equipe.

### Interpretação

1. Por que prever sempre ausência de fraude pode gerar acurácia alta?
2. Qual métrica descreve a parcela dos alertas emitidos que corresponde a fraudes reais?
