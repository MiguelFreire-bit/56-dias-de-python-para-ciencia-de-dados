## Dia 52 — Monitorar sem esconder as falhas

### Situação

Um serviço anuncia apenas a latência média das respostas bem-sucedidas. Essa comunicação deixa de mostrar erros, timeouts e respostas muito lentas. Você deverá produzir um resumo operacional que represente todas as tentativas e avalie duas metas didáticas.

### Dados disponíveis

Use `logs.csv`, com 20 requisições nas colunas `id`, `status`, `latencia_ms` e `tokens`. Os tokens faturados são exatamente os registrados. Um timeout de 1.000 ms indica o tempo observado até o corte, e não a duração completa de um processamento posterior desconhecido. As metas são no máximo 5% de falhas entre todas as tentativas e P95 de no máximo 500 ms entre respostas com status `ok`. Use quantis por interpolação linear e tarifa didática de R$ 0,01 por 1.000 tokens.

### Sua tarefa

1. Calcule as contagens e taxas de sucesso e falha. Considere erro e timeout como falhas e use todas as 20 tentativas no denominador.
2. Calcule média, mediana e percentil 95 das latências em dois recortes: somente sucessos e todas as tentativas. Identifique claramente cada recorte.
3. Avalie as duas metas com seus respectivos denominadores e subconjuntos.
4. Calcule o custo total e o custo por resposta bem-sucedida. O custo total deve considerar todos os tokens faturados, inclusive de falhas se houver. Recomende uma investigação prioritária.

### Entrega

Apresente um painel em tabela, uma visualização e um parecer operacional.

### Confira seu avanço

- [ ] A taxa de falha inclui erros e timeouts e usa 20 tentativas.
- [ ] O P95 informa recorte e convenção, e os custos não descartam tokens faturados.
- [ ] Não tratei o tempo do timeout como duração real do processamento depois do corte.

### Interpretação

1. Como excluir timeouts pode melhorar artificialmente a aparência do serviço?
2. Por que uma latência média baixa não assegura uma boa experiência para as respostas mais lentas?
