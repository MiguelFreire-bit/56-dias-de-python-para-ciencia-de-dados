## Dia 21 — Reamostrar para enxergar incerteza

### Situação

Uma equipe sorteou 12 atendimentos independentes de uma população ampla e mediu suas durações. Você deverá estimar o tempo médio e mostrar a incerteza dessa estimativa usando bootstrap. O conjunto é pequeno e assimétrico, de modo que as limitações do procedimento precisam acompanhar o resultado.

### Dados disponíveis

O arquivo `duracoes.csv` contém os tempos na coluna `minutos`. Para o bootstrap, gere 5.000 reamostras de tamanho 12, sorteando com reposição entre os tempos observados, com semente 210. O intervalo percentil de 95% será definido pelos quantis 2,5% e 97,5% das médias reamostradas, com interpolação linear.

### Sua tarefa

1. Calcule a média e o desvio padrão amostral dos 12 tempos originais.
2. Gere as reamostras e calcule uma média para cada uma. Faça um histograma das 5.000 médias.
3. Obtenha os dois limites do intervalo percentil e calcule o desvio padrão das médias bootstrap. Use `ddof=1` nas estimativas empíricas de desvio padrão.
4. Compare a dispersão dos tempos individuais com a dispersão das médias. Escreva a interpretação do intervalo e duas limitações do procedimento neste conjunto.

### Entrega

Inclua o código, o gráfico, a estimativa da média, o intervalo e uma nota sobre as limitações.

### Confira seu avanço

- [ ] Cada reamostra tem 12 observações selecionadas com reposição, produzindo 5.000 médias.
- [ ] O intervalo foi calculado com as médias bootstrap, e não com os percentis dos tempos individuais.
- [ ] Não atribuí 95% de probabilidade a um parâmetro fixo e reconheci que o intervalo é uma aproximação em amostra pequena.
