## Dia 40 — Um contrato de dados que detecta falhas

### Situação

Um processo de análise precisa impedir a entrada de registros inválidos, mas não pode descartar automaticamente toda ausência. Você deverá escrever uma validação com regras explícitas e testes que mostrem seu comportamento em casos diferentes.

### Dados disponíveis

Use `lote.csv`, com `id`, `idade`, `renda` e `segmento`. O contrato exige ID não vazio e único, idade inteira entre 0 e 120, renda numérica maior ou igual a zero ou ausente, e segmento A ou B. Renda ausente é permitida, mas gera aviso. Se um mesmo ID tiver conteúdos diferentes, todas as versões desse ID devem ir para quarentena. Neste exercício, trate também uma duplicação exata como violação de unicidade e registre as linhas para revisão, sem escolher uma silenciosamente.

### Sua tarefa

1. Implemente uma função que receba uma tabela e devolva registros válidos e um relatório de problemas. O relatório deve identificar a linha, a regra e se o problema é aviso ou impede o uso.
2. Preserve renda zero como valor válido e permita uma renda ausente quando nenhuma outra regra da linha for violada.
3. Identifique duplicações antes de selecionar as linhas válidas, mantendo todas as versões conflitantes fora da saída limpa.
4. Escreva testes com pequenos exemplos próprios para lote válido, idade inválida, renda ausente, segmento inválido e ID com conteúdos conflitantes. Verifique os resultados esperados e as mensagens.

### Entrega

Apresente função, saída válida, auditoria do arquivo e testes executáveis. O foco desta atividade é qualidade de dados e software; não há cálculo estatístico central obrigatório.

### Confira seu avanço

- [ ] Renda ausente gera aviso, enquanto zero continua válido; outras violações da mesma linha continuam sendo avaliadas.
- [ ] Nenhuma versão de um ID conflitante foi incluída silenciosamente entre os registros limpos.
- [ ] Os testes verificam o contrato e suas mensagens com expectativas explícitas, em vez de repetir a implementação.

### Interpretação

1. Que decisão de negócio seria escondida ao manter arbitrariamente a primeira versão de um ID?
2. Que definição precisaria ser acordada antes de preencher uma renda ausente?
