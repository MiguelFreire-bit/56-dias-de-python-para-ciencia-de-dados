## Dia 56 — Auditoria final: defender uma decisão

### Situação

Uma equipe recebeu previsões prontas de inflação e recomenda o modelo com menor erro. Antes de aceitar a recomendação, você deverá verificar se todos os candidatos poderiam realmente produzir suas previsões com a informação disponível na época. A qualidade do procedimento vem antes da ordenação das métricas.

### Dados disponíveis

Use `ipca_2023_2024.csv`, com o observado do IPCA/IBGE distribuído pelo Banco Central na série SGS 433; `previsoes_recebidas.csv`, com as previsões de `persistencia`, `linear` e `modelo_C`; e `metadados.json`, com as entradas e o procedimento de cada candidato. Os alvos são os 12 meses de 2024. Cada previsão deve usar somente informação disponível após a publicação da taxa do mês anterior. A unidade é variação percentual mensal.

### Sua tarefa

1. Leia integralmente os metadados e audite datas, unidades, duplicações, chaves e cobertura. Relacione previsões e observações pela data, sem depender da ordem das linhas.
2. Para cada candidato, verifique se suas entradas e seu ajuste respeitam a origem temporal da previsão. Defina quais candidatos são elegíveis e justifique qualquer exclusão antes de ordenar erros.
3. Escolha e justifique uma métrica principal. Compare os candidatos elegíveis nos mesmos 12 meses e examine os erros separadamente em cada semestre.
4. Produza um gráfico que ajude a avaliar os erros ou as previsões. Não treine modelos nem ajuste escolhas para melhorar os resultados desse período.
5. Escreva uma recomendação com as evidências da auditoria, os limites da comparação e uma avaliação futura que você pediria antes de recomendar uso real.

### Entrega

Apresente código de auditoria, tabela de elegibilidade, tabela de erros, gráfico e relatório de até 500 palavras. O relatório deve ser compreensível sem depender de uma explicação oral.

### Confira seu avanço

- [ ] As 12 datas de previsão foram conciliadas por chave com o observado, sem junção posicional silenciosa.
- [ ] Diferenciei erro baixo de procedimento válido, sustentando exclusões nos metadados.
- [ ] Não tratei os meses como observações independentes para criar um intervalo ingênuo nem fiz afirmações causais.
- [ ] Registrei recomendação, incerteza e próximo teste, mesmo que isso desfavoreça o candidato mais complexo.

### Interpretação

1. Que resultado aparentemente bom você rejeitaria por invalidade do procedimento?
2. O que você consegue defender com este período e o que precisaria observar em um período novo?
