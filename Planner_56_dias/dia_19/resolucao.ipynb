## Dia 19 — Por que médias variam menos

### Situação

Uma equipe observa que os tempos de atendimento são assimétricos e apresentam algumas esperas longas. Uma colega afirma que a média de 50 atendimentos varia tanto quanto o tempo de uma pessoa. Você deverá investigar essa afirmação por simulação, distinguindo a distribuição dos tempos da distribuição de suas médias.

### Dados disponíveis

O arquivo `parametros.csv` especifica uma distribuição exponencial com escala de 10 minutos, 2.000 repetições, tamanhos amostrais 5 e 50 e semente 190. Nesse modelo, a média e o desvio padrão populacionais são ambos 10 minutos. Não há uma população finita a ser esgotada: cada observação é uma nova realização independente da distribuição.

### Sua tarefa

1. Com `numpy.random.default_rng(190)`, gere 2.000 amostras de tamanho 5 e 2.000 de tamanho 50. Organize os resultados de forma que seja possível calcular a média dentro de cada amostra.
2. Construa três histogramas: um dos tempos individuais gerados nas amostras de tamanho 50, um das 2.000 médias de tamanho 5 e outro das 2.000 médias de tamanho 50. Identifique eixos e tamanhos amostrais.
3. Calcule o desvio padrão empírico das médias em cada tamanho e compare com `10 / sqrt(n)`. Declare o `ddof` usado na estimativa empírica.
4. Avalie a afirmação da colega com base nos gráficos e na tabela, explicando a diferença entre desvio padrão dos dados e erro padrão da média.

### Entrega

Apresente código, três histogramas, tabela de dispersão teórica e simulada e sua conclusão.

### Interpretação

1. Aumentar o tamanho das amostras faz os tempos individuais passarem a ter distribuição normal?
2. Que tipo de variação o erro padrão da média descreve?
