## Dia 1 — Antes de calcular, entender as colunas

### Situação

Uma loja reuniu informações de seis clientes atendidos durante uma semana. Antes de calcular médias ou comparar grupos, a responsável pela loja pediu que você explicasse o que cada coluna representa e quais operações fazem sentido com esses dados. Ela também quer saber quantos dos clientes registrados pertencem a cada bairro.

Sua análise descreve somente os seis clientes do arquivo. A loja tem outros clientes, que não foram observados nessa semana. Portanto, esse pequeno conjunto não representa um cadastro completo de toda a clientela.

### Dados disponíveis

Use o arquivo `clientes.csv`, que está na mesma pasta deste notebook. Cada linha representa um cliente. As cinco colunas são:

- `id`: código que identifica o cliente. Os zeros à esquerda fazem parte do código, como em `001`.
- `bairro`: bairro informado pelo cliente.
- `satisfacao`: avaliação do atendimento, com a ordem ruim < regular < boa < ótima.
- `compras`: quantidade de compras registradas para o cliente na semana observada.
- `gasto_reais`: valor total dessas compras, em reais, na mesma semana.

### Sua tarefa

1. Leia o CSV com Python ou pandas. Preserve as seis linhas e faça com que `id` seja lido como texto, mantendo os zeros à esquerda.
2. Monte um **dicionário de dados**, isto é, uma tabela que documente as cinco colunas. Para cada coluna, escreva seu significado, o tipo usado no código e sua classificação estatística. No tipo usado no código, informe como os valores foram armazenados, por exemplo, como texto ou número inteiro. Na classificação estatística, diferencie variáveis qualitativas nominais ou ordinais e quantitativas discretas ou contínuas. Justifique suas escolhas considerando o significado da coluna, e não apenas sua aparência.
3. Escolha uma das linhas e represente esse cliente como um **dicionário Python**, usando os nomes das colunas como chaves. Essa representação de um registro é diferente da tabela de documentação solicitada na etapa anterior.
4. Conte quantos clientes aparecem em cada bairro e apresente uma tabela com o bairro e sua contagem. Confira se o total das contagens corresponde ao número de linhas do arquivo.
5. Explique, em um pequeno parágrafo, qual é o grupo efetivamente observado e qual seria a população de interesse da loja caso ela quisesse conhecer toda a sua clientela. Diga por que esses seis registros, sozinhos, não garantem conclusões sobre todos os clientes.
