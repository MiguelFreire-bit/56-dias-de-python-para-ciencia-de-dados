## Dia 31 — A métrica muda a preferência

### Situação

Dois modelos já produziram previsões de demanda para os mesmos seis pedidos. A equipe quer escolher um deles, mas ainda não definiu como os erros afetam a operação. Você deverá comparar duas formas de penalização e mostrar como essa escolha pode mudar a preferência.

### Dados disponíveis

O arquivo `previsoes.csv` contém `real`, `modelo_A` e `modelo_B`, em unidades vendidas. Não treine modelos. No primeiro cenário, cada unidade de erro absoluto tem o mesmo custo. No segundo, a penalização cresce com o quadrado do erro, dando mais importância aos grandes desvios.

### Sua tarefa

1. Para cada modelo e pedido, calcule o erro assinado como previsto menos real, seu valor absoluto e seu quadrado.
2. Calcule MAE e RMSE usando os seis pedidos. O MAE é a média dos erros absolutos; o RMSE é a raiz quadrada da média dos erros quadráticos.
3. Escolha um modelo em cada cenário de custo e justifique a métrica usada.
4. Em uma análise adicional, refaça a comparação sem a última linha do arquivo. Mostre como a preferência ou a diferença entre modelos muda, mantendo a comparação com os seis casos como principal.

### Entrega

Apresente a tabela de erros por pedido, o resumo de métricas, a análise de sensibilidade e a escolha para cada cenário.

### Confira seu avanço

- [ ] A comparação principal usa os mesmos seis pedidos para os dois modelos.
- [ ] O RMSE aplica a raiz depois de calcular a média dos quadrados.
- [ ] Relacionei métrica e custo e não excluí um caso apenas para favorecer um modelo.
