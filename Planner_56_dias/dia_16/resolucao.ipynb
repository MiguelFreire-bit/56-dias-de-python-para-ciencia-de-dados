## Dia 16 — Do evento à distribuição

### Situação

Um sistema recebe solicitações em lotes de dez. Para planejar o atendimento, a equipe quer conhecer a chance de um lote apresentar pelo menos três falhas. Você deverá calcular essa chance segundo um modelo probabilístico e compará-la com uma simulação.

### Dados disponíveis

O arquivo `parametros.csv` define dez solicitações por lote, probabilidade de falha de 0,20 por solicitação, 10.000 lotes simulados e semente 160. Para este modelo didático, as solicitações são independentes e têm a mesma probabilidade de falha. Esses parâmetros são hipóteses do exercício, e não medições de um sistema real.

### Sua tarefa

1. Defina `X` como o número de falhas em um lote. Calcule teoricamente `P(X >= 3)` pela distribuição binomial, usando a soma das probabilidades de 3 até 10 falhas ou uma função de cauda equivalente.
2. Use `numpy.random.default_rng` com semente 160 para gerar 10.000 contagens binomiais com os parâmetros fornecidos.
3. Calcule a proporção dos lotes simulados que tiveram pelo menos três falhas e a diferença entre essa proporção e a probabilidade teórica.
4. Faça um gráfico das frequências do número de falhas por lote e explique por que a simulação não precisa coincidir exatamente com a teoria.

### Entrega

Inclua o código reproduzível, os resultados teórico e simulado, sua diferença e o gráfico.

### Confira seu avanço

- [ ] O evento inclui lotes com exatamente três falhas.
- [ ] A simulação tem 10.000 contagens entre 0 e 10 e pode ser repetida com a mesma semente.
- [ ] Quantifiquei a diferença entre os resultados e a interpretei como variação da simulação, sem exigir igualdade exata.

### Interpretação

1. Qual hipótese seria comprometida se uma falha aumentasse a chance de outras falhas no mesmo lote?
2. Aumentar o número de simulações corrigiria um modelo cuja hipótese de independência fosse falsa?
