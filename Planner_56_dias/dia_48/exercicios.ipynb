## Dia 48 — Um prompt melhor sob qual critério?

### Situação

Uma equipe comparou duas versões de prompt nas mesmas 12 perguntas. A versão B usa mais tokens, e a equipe quer saber se o resultado justifica seguir com ela. Você deverá analisar a comparação pareada, seu custo e a limitação de uma avaliação pequena.

### Dados disponíveis

O arquivo `prompts.csv` contém `caso`, `A_correto`, `B_correto`, `tokens_A` e `tokens_B`. Correto vale 1 e incorreto vale 0. Considere as 12 perguntas independentes e uma rubrica fixa. A tarifa didática é R$ 0,01 por 1.000 tokens, igual para ambas as versões. Não há nova execução de prompts nesta atividade.

### Sua tarefa

1. Monte uma tabela 2 × 2 com os resultados conjuntos de A e B, distinguindo concordâncias e discordâncias.
2. Calcule a taxa de acerto de cada versão, a diferença B menos A e os custos totais.
3. Defina `b` como o número de casos em que A acerta e B erra e `c` como o número em que A erra e B acerta. Faça teste binomial exato bilateral com `b` sucessos em `b+c` tentativas e probabilidade nula 0,5, usando nível 0,05. Se não houver discordâncias, registre esse fato sem inventar um teste.
4. Recomende um próximo passo considerando melhoria observada, evidência estatística, tamanho da avaliação e custo.

### Entrega

Apresente a tabela pareada, taxas, custos, teste e recomendação.

### Confira seu avanço

- [ ] Comparei os resultados como 12 pares, sem tratar os 24 julgamentos como independentes.
- [ ] O custo usa a divisão por 1.000 e a mesma tarifa para A e B.
- [ ] Diferenciei ganho observado, evidência estatística e relevância prática.

### Interpretação

1. Por que as concordâncias não entram como tentativas no teste binomial das discordâncias?
2. O que precisaria mudar na avaliação se as 12 perguntas fossem variações de apenas duas dúvidas?
