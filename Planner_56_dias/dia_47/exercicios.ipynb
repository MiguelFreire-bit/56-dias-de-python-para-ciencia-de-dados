## Dia 47 — JSON válido ainda pode estar errado

### Situação

Um extrator recebeu documentos e produziu respostas em JSON com prazo, custo e fonte. A equipe quer aprovar qualquer resposta que possa ser lida como JSON. Você deverá auditar separadamente a forma da saída, o contrato dos campos e sua sustentação no documento recebido.

### Dados disponíveis

Use `corpus.csv`, com `doc_id` e `texto`, e `saidas.csv`, com `resposta_id`, `documento_origem` e `json_recebido`. São seis saídas prontas; não execute um LLM. O contrato exige exatamente os campos `prazo_dias`, `custo_reais` e `fonte`. O prazo deve ser um inteiro não negativo ou `null`; o custo, um número não negativo ou `null`; e a fonte, o ID do documento recebido. Booleanos não contam como números. Não converta strings numéricas automaticamente.

O prazo representa apenas um prazo de devolução ou emissão. Uma quantidade de viagens por dia não é um prazo. Se uma informação não estiver no documento, não invente um valor; use `null`. Quando houver dias úteis, registre essa unidade em uma observação da auditoria, sem adicionar um campo ao JSON do contrato.

### Sua tarefa

1. Tente interpretar cada saída como JSON e registre se a leitura foi bem-sucedida.
2. Nas saídas interpretáveis, valide campos, tipos, limites e fonte de acordo com o contrato.
3. Compare cada saída com seu documento de origem, verificando significado, unidade e informações ausentes. Registre a evidência para cada aprovação ou rejeição.
4. Calcule três taxas sobre o total de seis saídas: leitura válida como JSON, conformidade com o contrato e conteúdo fundamentado. Para essas taxas, uma falha de leitura conta como falha nas três camadas. Uma saída legível pode ter falhas distintas nas outras camadas.

### Entrega

Apresente o validador, a auditoria por saída, as três taxas e a explicação de pelo menos duas falhas diferentes.

### Confira seu avanço

- [ ] Todas as seis saídas entram nos denominadores, inclusive a que não pode ser lida como JSON.
- [ ] Não aprovei o conteúdo apenas porque o formato ou os tipos estavam corretos.
- [ ] Cada rejeição aponta uma regra ou evidência documental, sem necessidade de chamada paga.
