## Dia 17 — Custo esperado não é custo garantido

### Situação

Uma operação precisa planejar uma reserva de recursos para problemas em pedidos. A gerência pediu o custo médio por pedido, mas também quer saber se esse número descreve bem o risco de um pedido isolado. Você deverá comparar o custo esperado com a dispersão e a ocorrência de eventos caros.

### Dados disponíveis

O arquivo `custos.csv` contém `tipo`, `probabilidade` e `custo_reais`. Cada pedido tem exatamente um dos três resultados: sem falha, falha leve ou falha grave. As probabilidades são 0,70, 0,25 e 0,05; os custos correspondentes são R$ 0, R$ 20 e R$ 300. Para a simulação, considere pedidos independentes.

### Sua tarefa

1. Confira se as probabilidades somam 1. Calcule o custo esperado `E(C)` e o valor esperado do quadrado do custo `E(C²)`.
2. Calcule a variância por `E(C²) - E(C)²` e obtenha o desvio padrão.
3. Simule 5.000 pedidos com semente 170. Compare a média simulada com o custo esperado e a proporção de custos acima de R$ 100 com a probabilidade teórica desse evento.
4. Escreva uma orientação para a gerência que diferencie o planejamento médio de muitos pedidos do risco de um pedido individual.
