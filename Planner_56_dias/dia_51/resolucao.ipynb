## Dia 51 — Mudança nos dados e no desempenho

### Situação

Uma equipe percebeu mudanças nas entradas de um serviço e concluiu que elas causaram uma piora no modelo. Você recebeu dados de duas janelas para separar três questões: se as entradas mudaram, se o desempenho mudou e o que é possível afirmar sobre a relação entre essas mudanças.

### Dados disponíveis

Use `janelas.csv`, com `janela`, `x`, `real` e `predito`. Cada janela tem dez casos independentes, e as duas amostras didáticas são distintas. `real` e `predito` estão na mesma unidade de demanda. Os rótulos reais já estão disponíveis, mas o arquivo não contém evidência sobre a causa de uma eventual mudança.

### Sua tarefa

1. Escolha resumos numéricos e uma visualização para comparar a distribuição de `x` entre as janelas. Explique por que essas escolhas ajudam a enxergar a mudança.
2. Calcule MAE em cada janela e compare o desempenho separadamente da mudança nas entradas.
3. Avalie a afirmação “detectei drift, logo o modelo piorou por causa do drift”, considerando o desenho e o pequeno número de casos.
4. Proponha um monitoramento que defina referência de comparação, janela, frequência, denominadores e uma regra de investigação. Trate seus limites de alerta como escolhas a validar, e não como valores universais.

### Entrega

Apresente gráfico, tabela de dados e desempenho e um plano breve de monitoramento.

### Confira seu avanço

- [ ] A comparação de entradas e a comparação de erros aparecem separadamente.
- [ ] Avaliei a conclusão causal sem atribuir à amostra evidência que ela não fornece.
- [ ] O plano define como medir e quando investigar, sem depender apenas de variações de média fora de contexto.

### Interpretação

1. Pode haver mudança nas entradas sem piora no erro? Explique.
2. Que parte da avaliação ainda seria possível se os rótulos reais recentes não estivessem disponíveis?
