## Dia 13 — Taxa de conversão em SQL

### Situação

Uma equipe de marketing compara dois canais usando a média simples de suas taxas diárias de compra. Os dias têm volumes diferentes de visitas, e um deles não recebeu visita alguma. Você deverá calcular os indicadores em SQL e explicar qual pergunta cada forma de resumo responde.

### Dados disponíveis

O arquivo `campanhas.csv` contém `dia`, `canal`, `visitas` e `compras`, com uma linha por combinação de dia e canal. Visitas e compras são contagens, e compras nunca supera visitas. Se não houve visitas, a taxa daquele dia é indefinida. Use SQLite pelo módulo `sqlite3`.

### Sua tarefa

1. Carregue o arquivo no SQLite e calcule a taxa de compra de cada dia e canal. Garanta divisão com resultado decimal e retorne `NULL` quando o número de visitas for zero.
2. Calcule a taxa global de cada canal dividindo a soma das compras pela soma das visitas.
3. Calcule, separadamente, a média simples das taxas diárias definidas. Mostre quantos dias entram nessa média.
4. Compare os dois indicadores e recomende qual deles usar para descrever a proporção de visitas que resultaram em compra no período inteiro.

### Entrega

Inclua as consultas, a tabela diária, a tabela comparativa por canal e sua recomendação. Você pode organizar a consulta com CTE ou subconsulta.

### Confira seu avanço

- [ ] A divisão preserva casas decimais, e o dia sem visitas recebe `NULL`, não taxa zero.
- [ ] A taxa global foi calculada como `SUM(compras) / SUM(visitas)`.
- [ ] Expliquei o efeito dos volumes diários e informei como o dia sem visitas foi tratado.

### Interpretação

1. Qual cálculo descreve a chance empírica de compra ao selecionar uma visita entre todas as visitas do período?
2. Por que a média simples das taxas pode levar a uma comparação diferente entre os canais?
