## Dia 24 — Comparar duas médias com hipóteses claras

### Situação

Uma equipe distribuiu pessoas aleatoriamente entre duas interfaces e mediu o tempo necessário para concluir uma tarefa. Cada pessoa utilizou uma única interface. Você deverá comparar as médias e avaliar se a diferença observada é relevante para a equipe.

### Dados disponíveis

O arquivo `grupos.csv` contém `grupo` e `minutos`, com oito pessoas em A e dez em B. Considere grupos independentes, tempos aproximadamente normais e variâncias populacionais desconhecidas, sem presumir que sejam iguais. Use teste t de Welch bilateral, com nível 0,05 e hipótese nula `mu_A - mu_B = 0`. Antes da análise, a equipe definiu como relevante uma redução de pelo menos dois minutos no tempo de B em relação a A; nessa direção, a diferença `A - B` deve ser pelo menos 2.

### Sua tarefa

1. Calcule tamanho, média e desvio padrão amostral de cada grupo.
2. Execute o teste de Welch e apresente a diferença `média_A - média_B`, mantendo essa ordem em todo o relatório.
3. Construa o intervalo de 95% da diferença com erro padrão `sqrt(sA²/nA + sB²/nB)` e graus de liberdade de Welch-Satterthwaite. Consulte a expressão dos graus de liberdade no material de apoio ou na documentação da biblioteca e registre a referência utilizada.
4. Avalie separadamente a evidência contra diferença nula e a evidência sobre uma redução de pelo menos dois minutos, considerando a largura do intervalo.

### Entrega

Apresente a tabela dos grupos, teste, diferença, intervalo e um parecer que separe significância estatística e relevância prática.

### Confira seu avanço

- [ ] Tratei os grupos como independentes e não criei pares entre pessoas diferentes.
- [ ] Usei variâncias amostrais e um procedimento que não exige variâncias iguais.
- [ ] Mantive o sinal de A menos B e considerei tanto a margem de dois minutos quanto a incerteza.
