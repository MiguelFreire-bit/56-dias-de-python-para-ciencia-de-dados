## Dia 54 — Complexidade precisa provar utilidade

### Situação

Uma equipe comparou uma arquitetura simples e outra multiagente nas mesmas tarefas, com repetições. Você recebeu os resultados simulados e deverá recomendar uma escolha considerando acerto, custo, latência e incerteza, sem tratar repetições da mesma tarefa como tarefas independentes.

### Dados disponíveis

O arquivo `comparacao.csv` contém 20 tarefas, com três rodadas cada, totalizando 60 linhas. A chave é `tarefa` mais `rodada`. As colunas de cada arquitetura informam acerto em 0/1, latência em milissegundos e custo em reais. Considere tarefas sorteadas independentemente da população-alvo; rodadas de uma mesma tarefa podem ser dependentes.

A regra didática foi definida antes da análise: adotar a arquitetura multiagente somente se o ganho médio de acerto for de pelo menos 0,05, ou cinco pontos percentuais, e suas médias de custo e latência não ultrapassarem o dobro das médias da simples. Cumprir essa regra pontual não elimina a necessidade de avaliar incerteza.

### Sua tarefa

1. Compare acerto, custo e latência das arquiteturas no total, por rodada e por tarefa.
2. Faça bootstrap pareado por tarefa: sorteie, com reposição, 20 IDs de tarefa e leve junto todas as três rodadas e ambas as arquiteturas de cada ID sorteado. Se um ID for sorteado mais de uma vez, seu grupo precisa aparecer com essa multiplicidade.
3. Gere 3.000 diferenças de acerto com semente 540 e calcule um intervalo percentil de 95%, com quantis lineares, para o ganho multiagente menos simples.
4. Aplique a regra de adoção e escreva uma recomendação de até 250 palavras que considere também o intervalo e o caráter simulado dos dados.

### Entrega

Apresente tabelas, código do bootstrap, intervalo do ganho e recomendação.

### Confira seu avanço

- [ ] Não mudei os limites de custo, latência ou ganho depois de observar os resultados.
- [ ] A recomendação admite manter a solução simples e diferencia simulação de evidência de produção.

### Interpretação

1. Por que escolher somente a melhor rodada pode favorecer indevidamente uma arquitetura?
2. Um ganho pontual acima do limite elimina a incerteza sobre tarefas novas?
