## Dia 22 — Um intervalo para a média

### Situação

Uma fábrica sorteou 12 embalagens de um processo estável para estimar o peso médio de produção. O desvio padrão populacional é desconhecido. Você deverá construir um intervalo de confiança e mostrar como a escolha do nível de confiança afeta sua largura.

### Dados disponíveis

Use a coluna `gramas` de `pesos.csv`. Para esta atividade, considere observações independentes e uma população aproximadamente normal. O valor de 500 g é uma referência de especificação, e não uma décima terceira observação. Use a distribuição t de Student, com `n-1` graus de liberdade.

### Sua tarefa

1. Calcule tamanho da amostra, média, desvio padrão amostral com `ddof=1` e erro padrão `s / sqrt(n)`.
2. Para confiança bilateral de 95%, obtenha o valor crítico da distribuição t e construa o intervalo como média mais ou menos valor crítico vezes erro padrão.
3. Confira o resultado com uma função da SciPy e construa também o intervalo bilateral de 90% para os mesmos dados.
4. Compare as larguras e explique a confiança como uma propriedade do procedimento em amostragens repetidas, indicando as hipóteses adotadas.

### Entrega
