## Dia 35 — Validação cruzada também tem limites

### Situação

Uma equipe comparou configurações de um classificador e quer divulgar a melhor média de validação cruzada como se fosse uma avaliação final independente. Você deverá organizar a seleção e mostrar a função diferente de um teste reservado.

### Dados disponíveis

Use `wine.csv`, com 178 amostras da base Wine/UCI, 13 medidas químicas, `row_id` e `target` em 0, 1 e 2. Separe o alvo e exclua o identificador das entradas. Reserve 25% como teste estratificado, com semente 35. Chame os 75% restantes de conjunto de desenvolvimento.

### Sua tarefa

1. No desenvolvimento, crie cinco divisões estratificadas com `StratifiedKFold(5, shuffle=True, random_state=35)`. Use as mesmas divisões para todas as configurações.
2. Compare KNN com 3, 5 e 9 vizinhos por macro-F1. Em cada divisão, ajuste a padronização somente na parcela de treino, por meio de um `Pipeline`.
3. Apresente as cinco pontuações de cada configuração, sua média e seu desvio padrão amostral. Escolha a maior média; em empate até `1e-12`, prefira o maior número de vizinhos.
4. Ajuste a configuração escolhida em todo o desenvolvimento e avalie uma única vez no teste. Explique por que o desvio entre divisões não é automaticamente um intervalo de confiança da generalização.

### Entrega

Inclua a tabela por divisão, os resumos, a decisão registrada antes do teste e o resultado final.

### Confira seu avanço

- [ ] O padronizador é reajustado dentro de cada divisão de validação cruzada.
- [ ] O teste ficou fora da seleção e foi usado somente na avaliação final.
- [ ] Descrevi a variação entre divisões sem tratá-las como observações independentes para construir um intervalo automático.

### Interpretação

1. Por que escolher a configuração de maior média e divulgar apenas essa média pode produzir uma avaliação otimista?
2. Repetir os cortes até obter um teste melhor preservaria a independência da avaliação final?
