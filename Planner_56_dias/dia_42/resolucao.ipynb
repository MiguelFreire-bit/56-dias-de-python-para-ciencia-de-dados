## Dia 42 — Diagnosticar uma média no serviço

### Situação

Uma API promete informar o gasto médio por cliente cadastrado, incluindo quem não comprou no período. A consulta atual calcula a média dos valores de pedidos depois de uma junção que elimina clientes sem compra. Você deverá corrigir o indicador e proteger seu significado com testes.

### Dados disponíveis

Use `clientes.csv`, com `cliente` e `regiao`, `pedidos.csv`, com `cliente` e `valor`, e `consulta_com_erro.sql`. Há três clientes cadastrados: A no Norte e B e C no Sul. Todos os pedidos do período estão presentes. Para um cliente cadastrado sem pedido, o gasto nesse período é zero. Uma região que não tem clientes cadastrados é outra situação e deve gerar erro explícito.

### Sua tarefa

1. Execute ou examine a consulta com erro e explique dois problemas: qual unidade sua média está resumindo e quais clientes ficam de fora.
2. Escreva uma consulta que obtenha primeiro o total por cliente e mantenha todos os cadastrados. Depois calcule, por região, a quantidade de clientes e seu gasto médio.
3. Implemente uma função local que receba o nome da região e devolva quantidade de clientes e gasto médio. Use SQLite; não é necessário subir um servidor HTTP.
4. Escreva testes para Norte, Sul e uma região inexistente, usando resultados esperados que você verificou a partir dos registros.

### Entrega

Apresente SQL corrigido, função, testes e uma explicação do indicador antes e depois da correção.

### Confira seu avanço

- [ ] Agreguei os pedidos por cliente antes da média e incluí clientes sem pedidos.
- [ ] A contagem representa clientes do cadastro, sem multiplicá-los pelo número de pedidos.
- [ ] Os testes diferenciam uma região sem cadastro de uma região com clientes cujo gasto é zero.

### Interpretação

1. Por que média por pedido e média por cliente respondem a perguntas diferentes?
2. Por que zero é adequado para o gasto sem pedidos neste período, mas não para toda ausência de informação?
