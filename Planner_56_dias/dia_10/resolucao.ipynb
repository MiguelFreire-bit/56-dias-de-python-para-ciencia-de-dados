## Dia 10 — Quando o agregado muda a história

### Situação

Uma equipe está comparando dois procedimentos fictícios, A e B, pela taxa de sucesso. Entretanto, os procedimentos não receberam a mesma composição de casos leves e graves. Você precisa verificar se a comparação geral esconde diferenças importantes entre esses grupos.

### Dados disponíveis

Use `tratamentos.csv`, com as colunas `grupo`, `gravidade`, `sucessos` e `total`. Cada linha descreve um grupo de pessoas diferente, sem sobreposição com as demais linhas. O número de falhas é a diferença entre total e sucessos. As pessoas não foram alocadas aleatoriamente aos procedimentos. O contexto é inteiramente fictício.

### Sua tarefa

1. Calcule a taxa de sucesso de A e B separadamente para os casos leves e graves, dividindo sucessos pelo total correspondente.
2. Calcule a taxa geral de cada procedimento a partir das somas de sucessos e de pessoas, sem tirar a média simples das taxas das linhas.
3. Mostre a composição dos procedimentos: qual proporção dos casos de cada um é leve e qual é grave?
4. Construa um gráfico que permita comparar os procedimentos dentro de cada nível de gravidade. Escreva uma avaliação sobre o uso da taxa geral para recomendar um dos procedimentos.

### Entrega

Inclua as tabelas de taxas por gravidade e gerais, a tabela de composição, o gráfico e a avaliação da recomendação.

### Confira seu avanço

- [ ] Todas as taxas usam os números de sucessos e pessoas correspondentes ao grupo analisado.
- [ ] Quantifiquei as diferenças de composição e as relacionei ao resultado agregado.
- [ ] Minha conclusão distingue associação observada de superioridade causal de um procedimento.

### Interpretação

1. Como as comparações por gravidade e a comparação geral podem sugerir ordens diferentes entre A e B?
2. O que a alocação aleatória acrescentaria ao desenho dessa comparação?

### Leituras de apoio

Fávero e Belfiore, Manual de Análise de Dados, caps. 3-4; Análise de Dados: Da Teoria à Prática, p. 46. Contexto inteiramente fictício.
