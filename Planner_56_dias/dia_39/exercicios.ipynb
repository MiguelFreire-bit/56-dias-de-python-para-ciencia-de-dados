## Dia 39 — Comparar modelos preservando os pares

### Situação

Dois modelos foram avaliados nos mesmos 12 casos, sorteados independentemente da população de interesse. A equipe quer saber se a diferença dos erros médios oferece evidência suficiente para preferir um deles. Você deverá acompanhar essa diferença com uma medida de incerteza, preservando a comparação caso a caso.

### Dados disponíveis

O arquivo `erros.csv` contém `caso`, `erro_abs_A` e `erro_abs_B`. Os erros absolutos já foram calculados e usam a mesma unidade. Defina `delta = MAE_A - MAE_B`. Faça bootstrap pareado com 5.000 reamostras dos 12 casos, com reposição e semente 390. Use intervalo percentil de 95%, com quantis lineares.

### Sua tarefa

1. Calcule o MAE de cada modelo e a diferença delta. Explique qual modelo é favorecido por um delta positivo ou negativo.
2. Em cada reamostra, selecione índices de casos e aplique a mesma seleção às colunas de A e B. Calcule um delta para cada reamostra.
3. Obtenha o intervalo pelos percentis 2,5% e 97,5% dos deltas reamostrados.
4. Compare sua conclusão com a simples ordenação dos dois MAEs e explique o que precisaria mudar se os casos fossem dias consecutivos dependentes.

### Entrega
