## Dia 43 — Representar texto exige escolhas

### Situação

Um portal reúne informações sobre biblioteca, transporte e cursos. Antes de implementar uma busca, a equipe quer conhecer o vocabulário do acervo. Você deverá separar os textos em unidades de contagem e mostrar como escolhas de normalização alteram os resultados.

### Dados disponíveis

O arquivo `corpus.csv` contém oito documentos fictícios nas colunas `doc_id`, `tema` e `texto`. Cada linha representa um documento. Não é necessário baixar um modelo de linguagem. Nesta atividade, um token será uma sequência contínua de letras, depois da conversão do texto para minúsculas.

### Sua tarefa

1. Implemente uma tokenização que reconheça letras Unicode, preservando acentos. Números e sinais de pontuação não devem ser tokens nessa regra.
2. Crie uma segunda versão que remova os acentos antes da mesma tokenização. Preserve o texto original e mantenha a associação de cada lista de tokens com seu documento.
3. Escolha cinco termos, incluindo `cartão` e `Python`. Na versão sem acentos, consulte as formas normalizadas correspondentes.
4. Para cada termo e versão, calcule a frequência total de ocorrências e o número de documentos em que ele aparece. Explique uma consequência concreta da escolha de normalização, usando um exemplo do acervo.

### Entrega

Apresente as funções, as duas representações tokenizadas, a tabela dos cinco termos e a justificativa das regras.

### Confira seu avanço

- [ ] Os documentos mantêm seus IDs e textos originais.
- [ ] A frequência de documentos conta cada documento no máximo uma vez por termo, mesmo que ele se repita no texto.
- [ ] Registrei a regra e demonstrei como ela modifica uma representação ou uma consulta.

### Interpretação

1. Como remover acentos pode ajudar uma busca e, em outros casos, eliminar distinções úteis?
2. Por que essa tokenização por palavras não deve ser tratada como equivalente à tokenização de um LLM?
