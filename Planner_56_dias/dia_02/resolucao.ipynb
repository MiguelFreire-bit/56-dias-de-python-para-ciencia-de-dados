## Dia 2 — Contar com o denominador certo

### Situação

Uma equipe de atendimento recebeu 12 avaliações de satisfação e quer apresentar os resultados em uma reunião. Sua tarefa é mostrar quantas respostas existem em cada categoria e qual parcela dos respondentes avaliou o atendimento como bom ou ótimo. A apresentação deve preservar a ordem das avaliações sem tratá-las como uma escala de distâncias numéricas iguais.

### Dados disponíveis

O arquivo `respostas.csv` contém uma coluna chamada `resposta`, com uma avaliação válida por linha. Use a ordem ruim < regular < boa < ótima. As 12 linhas são respostas recebidas, e não o total de pessoas convidadas a responder.

### Sua tarefa

1. Leia o arquivo e conte as respostas em cada categoria, respeitando a ordem indicada.
2. Monte uma tabela com a frequência absoluta, a frequência relativa e a frequência relativa acumulada. A frequência absoluta é a contagem. A relativa é a parcela das 12 respostas naquela categoria. A acumulada reúne as parcelas desde a primeira categoria até a categoria da linha.
3. Faça um gráfico de barras com as categorias e suas contagens. Identifique os eixos e informe o total de respostas.
4. Calcule a proporção de avaliações boas ou ótimas e escreva um parágrafo curto para comunicar o resultado à equipe.

### Entrega

Apresente o código, a tabela de frequências, o gráfico e o parágrafo de interpretação.

### Confira seu avanço

- [ ] As contagens somam 12, e as frequências relativas somam 1 antes do arredondamento.
- [ ] A frequência acumulada segue a ordem de satisfação e termina em 1.
- [ ] O gráfico mostra contagens e não sugere que a distância entre duas categorias seja uma quantidade conhecida.

### Interpretação

1. Se houvesse pessoas convidadas que não responderam, o que mudaria ao incluí-las no denominador? Você ainda estaria medindo a satisfação dos respondentes?
2. Por que ordenar as categorias alfabeticamente prejudicaria a interpretação da frequência acumulada?
