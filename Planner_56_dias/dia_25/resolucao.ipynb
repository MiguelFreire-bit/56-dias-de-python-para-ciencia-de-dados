## Dia 25 — Antes e depois da mesma pessoa

### Situação

Oito pessoas executaram uma tarefa antes e depois de um treinamento. A equipe quer avaliar a mudança no tempo médio, preservando a correspondência entre as duas medições de cada participante. Como não houve grupo de controle, a conclusão também precisa reconhecer os limites do desenho.

### Dados disponíveis

Use `pares.csv`, com as colunas `pessoa`, `antes` e `depois`, em minutos. As duas medidas da mesma linha pertencem à mesma pessoa. Considere as diferenças aproximadamente normais e independentes entre participantes. Defina diferença como depois menos antes e use teste bilateral com nível 0,05.

### Sua tarefa

1. Calcule a diferença de cada pessoa e faça um gráfico dos pares ou das diferenças, mantendo os identificadores.
2. Calcule a média e o desvio padrão amostral das oito diferenças.
3. Execute o teste t pareado e construa um intervalo t de 95% para a diferença média.
4. Interprete a direção da mudança. Explique por que tratar as duas colunas como amostras independentes perderia informação e por que a mudança observada não comprova, sozinha, efeito causal do treinamento.

### Entrega
