## Dia 14 — Um gráfico para cada pergunta

### Situação

Um relatório sobre inflação usa apenas um histograma para explicar em quais meses as taxas subiram ou caíram. Além disso, chama a soma das taxas mensais de inflação acumulada anual. Você deverá corrigir a apresentação, escolhendo gráficos e cálculos adequados a cada pergunta.

### Dados disponíveis

Use `ipca_2023_2024.csv`, com 24 observações nas colunas `data` e `ipca_pct`, de janeiro de 2023 a dezembro de 2024. São taxas mensais do IPCA produzido pelo IBGE e distribuído pelo Banco Central, série SGS 433. A cópia local contém tudo o que você precisa.

### Sua tarefa

1. Ordene os registros por data e produza um gráfico de linhas para mostrar a evolução das taxas mensais.
2. Construa um histograma com os mesmos 24 valores. Informe os limites das classes e explique que o gráfico resume a distribuição das taxas, sem preservar a sequência dos meses.
3. Calcule a inflação acumulada de cada ano por composição: converta cada taxa mensal em fator `1 + taxa/100`, multiplique os 12 fatores e converta o resultado novamente em porcentagem.
4. Calcule a soma das taxas apenas para comparação com o acumulado composto. Apresente a diferença e explique por que os dois cálculos não são equivalentes.

### Entrega

Apresente os dois gráficos, a tabela anual dos cálculos e um texto indicando a pergunta respondida por cada representação.

### Confira seu avanço

- [ ] O gráfico temporal tem datas ordenadas, e o histograma usa os mesmos 24 meses.
- [ ] O acumulado anual usa o produto dos fatores e foi convertido novamente para porcentagem.
- [ ] Os eixos e as classes estão identificados, e não interpretei o histograma como uma sequência temporal.

### Interpretação

1. Que informação fica mais difícil de enxergar em cada um dos dois gráficos?
2. Por que as taxas negativas presentes neste arquivo não produzem fatores de composição negativos?

### Leituras de apoio

Fávero e Belfiore, Manual de Análise de Dados, cap. 2; Análise de Dados: Da Teoria à Prática, pp. 17-21; complemento de composição de taxas.
