## Dia 7 — A média das médias pode enganar

### Situação

Uma coordenação escolar quer divulgar a nota média dos alunos de três turmas. Alguém sugeriu calcular a média simples das três médias de turma. Como as turmas têm números diferentes de alunos, você deverá verificar qual cálculo responde à pergunta da coordenação.

### Dados disponíveis

O arquivo `turmas.csv` tem as colunas `turma`, `alunos` e `media_nota`. A turma A tem 10 alunos e média 9; a B tem 30 alunos e média 6; a C tem 60 alunos e média 7. Cada aluno fez uma prova e pertence a uma única turma. As médias fornecidas são exatas, mas as notas individuais não estão disponíveis.

### Sua tarefa

1. Calcule a média simples das três médias de turma.
2. Calcule a média ponderada pelo número de alunos, mostrando os pesos utilizados e o total de alunos.
3. Associe cada resultado à pergunta que ele responde: a nota média de um aluno escolhido entre todos os alunos ou a média de uma turma escolhida com igual chance entre as três.
4. Recomende o cálculo adequado para divulgar a nota média dos alunos e explique por que os dois resultados podem diferir.
