## Resultados e interpretação

### 1. Que informação adicional seria necessária para calcular o desvio padrão das notas individuais?

Seria necessário dispor das **notas individuais de todos os 100 alunos** ou, alternativamente, do **desvio padrão (ou variância) interno de cada turma**.

Para esse segundo caso, aplica-se a **Lei da Variância Total**, que decompõe a variação global em duas parcelas: a variação **entre** as médias das turmas (já calculável com os dados atuais) e a variação **dentro** de cada turma (a informação ausente):

$$\sigma_{\text{total}}^2 = \underbrace{\frac{\sum n_i \cdot \sigma_i^2}{N}}_{\text{dentro das turmas}} + \underbrace{\frac{\sum n_i \cdot (\bar{x}_i - \bar{x}_{\text{global}})^2}{N}}_{\text{entre as turmas}}$$

#### Exemplo ilustrativo com os dados do exercício
Considerando a Turma A ($n = 10$, média $9$), Turma B ($n = 30$, média $6$), Turma C ($n = 60$, média $7$) e média global $6{,}9$:

1. **Variação entre turmas (já calculável):**
   $$\frac{10(9 - 6{,}9)^2 + 30(6 - 6{,}9)^2 + 60(7 - 6{,}9)^2}{100} = \frac{44{,}1 + 24{,}3 + 0{,}6}{100} = 0{,}69$$

2. **Variação dentro das turmas (hipotética):** supondo que a coordenação informasse os desvios padrões internos $\sigma_A = 1$, $\sigma_B = 2$ e $\sigma_C = 1$, calcula-se a média ponderada das variâncias:
   $$\frac{10 \cdot 1^2 + 30 \cdot 2^2 + 60 \cdot 1^2}{100} = \frac{10 + 120 + 60}{100} = 1{,}90$$

3. **Desvio padrão global exato:** soma-se as duas parcelas ($1{,}90 + 0{,}69 = 2{,}59$) e extrai-se a raiz quadrada:
   $$\sigma_{\text{total}} = \sqrt{2{,}59} \approx 1{,}61$$

> **Conclusão:** Mesmo sem acesso às 100 notas individuais, conhecer o desvio padrão interno de cada turma já permitiria reconstruir o desvio padrão geral exato.

Os dois cálculos resultariam no mesmo valor em qualquer uma das seguintes situações:

- **Médias iguais:** quando todas as turmas apresentam a mesma nota média, independentemente da quantidade de alunos em cada uma.
