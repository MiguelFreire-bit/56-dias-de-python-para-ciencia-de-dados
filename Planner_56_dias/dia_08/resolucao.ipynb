## Dia 8 — Primeira análise de uma série oficial

### Situação

Uma equipe está preparando um relatório sobre a inflação mensal no Brasil e quer comparar 2023 com 2024. Sua tarefa é descrever o comportamento das taxas em cada ano, considerando tanto os valores centrais quanto sua dispersão. O relatório não deve confundir uma taxa mensal típica com a inflação acumulada no ano.

### Dados disponíveis

Use a cópia local `ipca_2023_2024.csv`. Ela contém 24 observações mensais, de janeiro de 2023 a dezembro de 2024, nas colunas `data` e `ipca_pct`. Os dados são do IPCA produzido pelo IBGE e distribuído pelo Banco Central na série SGS 433. `ipca_pct` é a variação percentual de cada mês. Não é necessário acessar a internet.

### Sua tarefa

1. Leia o arquivo, converta `data` para um tipo de data e verifique se `ipca_pct` está numérico.
2. Confira se há exatamente um registro para cada mês esperado, sem duplicatas ou meses ausentes. Separe as observações por ano.
3. Calcule, para cada ano, média, mediana, mínimo, máximo e desvio padrão com `ddof=0`. Esse divisor será usado para descrever todos os meses do recorte anual fornecido.
4. Escreva uma comparação dos anos usando os resultados da tabela. Descreva o que aconteceu no período observado, sem atribuir causas nem apresentar conclusões sobre outros períodos.

### Entrega

Inclua as verificações da base, a tabela anual e um texto de até oito linhas comparando os anos.

### Confira seu avanço

- [ ] Cada ano tem 12 meses únicos, e as taxas estão armazenadas como números.
- [ ] Identifiquei as medidas como resumos das taxas mensais, sem chamá-las de inflação acumulada anual.
- [ ] Minha comparação considera centro e dispersão e não atribui causalidade às mudanças.

### Interpretação

1. Como você explicaria uma taxa mensal negativa a uma pessoa que está lendo o relatório?
2. A média das taxas mensais responde à mesma pergunta que o acumulado anual? Explique a diferença.

### Leituras de apoio

Fávero e Belfiore, Manual de Análise de Dados, cap. 2; Análise de Dados: Da Teoria à Prática, pp. 13-16.
