## Dia 34 — Complexidade e overfitting

### Situação

Uma equipe quer escolher a profundidade de uma árvore de classificação. Uma árvore mais complexa pode memorizar o treino sem melhorar a previsão de novas plantas. Você deverá escolher a complexidade pela validação e reservar o teste para uma única avaliação final.

### Dados disponíveis

Use `iris.csv`, com 150 plantas da base Iris/UCI, quatro medidas em centímetros, `row_id` e `target`. As classes 0, 1 e 2 representam as três espécies. Exclua identificador e alvo das entradas. A atividade usa três conjuntos distintos: treino, validação e teste.

### Sua tarefa

1. Reserve 20% dos dados como teste, com estratificação e semente 34. Dos 80% restantes, reserve 25% para validação, também estratificada, com semente 35. Confira se não há sobreposição de índices.
2. Treine árvores com `max_depth` igual a 1, 2, 4 e `None`, todas com `random_state=34`. Compare macro-F1 no treino e na validação sem consultar o teste.
3. Escolha a maior macro-F1 de validação. Se a diferença entre as melhores pontuações for de até `1e-12`, prefira a menor profundidade; considere `None` a opção mais profunda. Registre a escolha antes da avaliação final.
4. Reajuste a árvore escolhida usando treino e validação juntos. Avalie uma única vez no teste e discuta o padrão de desempenho das profundidades.

### Entrega

Apresente a tabela ou curva de complexidade, a regra de escolha, o registro da decisão e o resultado final no teste.

### Confira seu avanço

- [ ] Os três conjuntos são distintos, e o teste não participou da escolha.
- [ ] Há pontuações de treino e validação para as quatro profundidades e somente uma avaliação final de teste.
- [ ] Discuti overfitting com base nos resultados, sem exigir que uma profundidade específica necessariamente perca.
