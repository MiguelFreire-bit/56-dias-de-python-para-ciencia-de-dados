## Dia 28 — Não detectar não é provar ausência

### Situação

Uma equipe quer saber se um teste consegue detectar o aumento de um minuto no tempo médio de atendimento. Em vez de analisar apenas um conjunto, você deverá simular muitos estudos em cenários cujo efeito verdadeiro é conhecido e comparar dois tamanhos amostrais.

### Dados disponíveis

O arquivo `parametros.csv` define média 10 sob a hipótese nula, média 11 no cenário com efeito, desvio padrão populacional 3, tamanhos 15 e 60, nível 0,05, 1.000 repetições por cenário e semente 280. As observações simuladas são normais e independentes. Em cada repetição, use um teste t bilateral contra média 10, estimando o desvio padrão a partir da própria amostra.

### Sua tarefa

1. Organize quatro cenários: cada combinação de média verdadeira 10 ou 11 com tamanho 15 ou 60.
2. Com um gerador de semente 280, simule 1.000 amostras em cada cenário. Execute o teste em cada amostra e registre se rejeitou a hipótese nula.
3. Calcule a taxa de rejeição de cada cenário. Quando a média verdadeira for 10, interprete-a como estimativa do erro tipo I; quando for 11, interprete-a como estimativa do poder. Calcule também `beta = 1 - poder`.
4. Compare os tamanhos e explique por que as taxas simuladas variam, sem exigir que o erro tipo I observado seja exatamente 0,05.

### Entrega

Apresente código reproduzível, tabela dos quatro cenários e uma recomendação sobre tamanho amostral e interpretação de resultados não significativos.
