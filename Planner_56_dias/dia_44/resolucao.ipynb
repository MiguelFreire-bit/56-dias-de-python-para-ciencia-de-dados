## Dia 44 — Uma busca lexical para servir de referência

### Situação

Um portal quer encontrar documentos que respondam às perguntas dos usuários. Antes de adicionar modelos mais complexos, você deverá construir uma busca lexical simples e inspecionável. Ela será uma referência para conhecer acertos e limitações do acervo.

### Dados disponíveis

Use `corpus.csv`, com oito documentos fictícios em `doc_id`, `tema` e `texto`, e `consultas.csv`, com quatro perguntas em `id`, `pergunta` e `doc_relevante`. O documento relevante serve apenas para avaliar o resultado; ele não deve participar do cálculo da similaridade. Tudo deve funcionar localmente.

### Sua tarefa

1. Ajuste `TfidfVectorizer(lowercase=True, strip_accents="unicode")` somente nos textos dos documentos.
2. Transforme as perguntas com o vetorizador já ajustado, sem reaprender o vocabulário com elas.
3. Calcule similaridade de cosseno e implemente uma função que devolva os três documentos de maior score para uma pergunta. Em empate, ordene por `doc_id` crescente, inclusive quando todos os scores forem zero.
4. Apresente o ranking das quatro perguntas e compare com os documentos relevantes. Analise pelo menos um acerto e uma limitação, sem modificar os documentos para esconder falhas.

### Entrega

Apresente a função de busca, rankings com IDs e scores e uma análise dos resultados. O foco é recuperação de documentos; não é necessário gerar respostas com LLM.

### Confira seu avanço

- [ ] Cada ranking contém três IDs distintos, com scores em ordem não crescente e desempate declarado.
- [ ] As perguntas não entraram no ajuste do vocabulário, e os rótulos relevantes não influenciaram a similaridade.
- [ ] Descrevi o score como similaridade, sem chamá-lo de probabilidade de a resposta estar correta.

### Interpretação

1. Por que uma pergunta que usa outras palavras pode falhar mesmo quando a informação existe no acervo?
2. Que limitações essa referência simples permite identificar antes de adicionar embeddings ou RAG?
