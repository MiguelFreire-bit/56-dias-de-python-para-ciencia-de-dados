## Dia 50 — O agregado esconde quem recebe o erro

### Situação

Um classificador identifica usuários que precisam de suporte, tanto em aplicativo quanto na web. O relatório apresenta apenas um indicador global, embora os canais tenham tamanhos e proporções de positivos diferentes. Você deverá examinar os erros por canal antes de recomendar mudanças.

### Dados disponíveis

O arquivo `segmentos.csv` contém `canal`, `real`, `predito` e `n`. A classe positiva, 1, significa precisar de suporte. Cada linha é uma combinação de resultados com frequência `n`; as pessoas não se repetem entre linhas. Os dados não garantem representatividade de todos os usuários futuros.

### Sua tarefa

1. Reconstrua as contagens da matriz de confusão de cada canal usando `n` como peso e confira o total agregado.
2. Calcule, por canal e no total, pelo menos recall, taxa de falsos positivos e suporte de classe. Suporte é a quantidade de casos reais de cada classe; os denominadores de recall e falsos positivos devem ficar explícitos.
3. Acrescente outra métrica se ela ajudar sua análise e faça uma visualização que permita comparar os canais, informando seus tamanhos.
4. Escreva um parecer de até 250 palavras com uma investigação recomendada antes de alterar limiares ou prometer equidade entre grupos.

### Entrega

Apresente a tabela estratificada e global, a visualização e o parecer.

### Confira seu avanço

- [ ] As contagens consideram `n`, e a soma dos canais confere com o total.
- [ ] Recall e taxa de falsos positivos usam os denominadores correspondentes às classes reais positivas e negativas.
- [ ] Diferenciei disparidade observada, incerteza e causas ainda não demonstradas, sem supor que igualar taxas resolve o problema.

### Interpretação

1. Um bom resultado global assegura bom atendimento em cada canal?
2. Que informação adicional ajudaria a investigar por que os erros diferem?
